````python
# =============================================================================
# CUSTOM MODULE REGISTRATION (MOVED TO TOP - ALWAYS RUN FIRST)
# =============================================================================

from ultralytics.nn.modules import C2f, Detect
import torch.nn as nn
import sys

# --- ECA Attention Block ---
class ECAAttention(nn.Module):
    def __init__(self, channels, k_size=3):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv = nn.Conv1d(1, 1, kernel_size=k_size, padding=k_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        y = self.avg_pool(x)
        y = self.conv(y.squeeze(-1).transpose(-1, -2)).transpose(-1, -2).unsqueeze(-1)
        y = self.sigmoid(y)
        return x * y

# --- CBAM Block ---
class ChannelAttention(nn.Module):
    def __init__(self, in_planes, ratio=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.max_pool = nn.AdaptiveMaxPool2d(1)
        self.fc = nn.Sequential(
            nn.Conv2d(in_planes, in_planes // ratio, 1, bias=False),
            nn.ReLU(),
            nn.Conv2d(in_planes // ratio, in_planes, 1, bias=False)
        )
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = self.fc(self.avg_pool(x))
        max_out = self.fc(self.max_pool(x))
        out = avg_out + max_out
        return self.sigmoid(out)

class SpatialAttention(nn.Module):
    def __init__(self, kernel_size=7):
        super().__init__()
        self.conv = nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    def forward(self, x):
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        x = torch.cat([avg_out, max_out], dim=1)
        x = self.conv(x)
        return self.sigmoid(x)

class CBAM(nn.Module):
    def __init__(self, channels, ratio=16, kernel_size=7):
        super().__init__()
        self.ca = ChannelAttention(channels, ratio)
        self.sa = SpatialAttention(kernel_size)
    def forward(self, x):
        x = x * self.ca(x)
        x = x * self.sa(x)
        return x

# --- SE Block ---
class SEBlock(nn.Module):
    def __init__(self, channels, reduction=16):
        super().__init__()
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, channels // reduction, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(channels // reduction, channels, bias=False),
            nn.Sigmoid()
        )
    def forward(self, x):
        b, c, _, _ = x.size()
        y = self.avg_pool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)

# --- DropBlock2D ---
class DropBlock2D(nn.Module):
    def __init__(self, drop_prob=0.1, block_size=7):
        super().__init__()
        self.drop_prob = drop_prob
        self.block_size = block_size
    def forward(self, x):
        if not self.training or self.drop_prob == 0.0:
            return x
        gamma = self.drop_prob / (self.block_size ** 2)
        mask = (torch.rand(x.shape[0], 1, x.shape[2], x.shape[3], device=x.device) < gamma).float()
        mask = nn.functional.max_pool2d(mask, self.block_size, stride=1, padding=self.block_size // 2)
        mask = 1 - mask
        return x * mask * (mask.numel() / mask.sum())

# --- Focal Loss ---
class FocalLoss(nn.Module):
    def __init__(self, alpha=1, gamma=2, reduction='mean'):
        super().__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction
    def forward(self, inputs, targets):
        BCE_loss = nn.functional.binary_cross_entropy_with_logits(inputs, targets, reduction='none')
        pt = torch.exp(-BCE_loss)
        F_loss = self.alpha * (1-pt)**self.gamma * BCE_loss
        if self.reduction == 'mean':
            return F_loss.mean()
        elif self.reduction == 'sum':
            return F_loss.sum()
        else:
            return F_loss

# --- Custom C2f and Detect_Enhanced ---
class C2f_ECA(C2f):
    def __init__(self, c1, c2, n=1, shortcut=False, g=1, e=0.5):
        super().__init__(c1, c2, n, shortcut, g, e)
        self.attention = ECAAttention(c2)
    def forward(self, x):
        return self.attention(super().forward(x))

class C2f_CBAM(C2f):
    def __init__(self, c1, c2, n=1, shortcut=False, g=1, e=0.5):
        super().__init__(c1, c2, n, shortcut, g, e)
        self.attention = CBAM(c2)
    def forward(self, x):
        return self.attention(super().forward(x))

class C2f_SE(C2f):
    def __init__(self, c1, c2, n=1, shortcut=False, g=1, e=0.5):
        super().__init__(c1, c2, n, shortcut, g, e)
        self.attention = SEBlock(c2)
    def forward(self, x):
        return self.attention(super().forward(x))

class C2f_Attention(C2f_CBAM):
    pass

class Detect_Enhanced(Detect):
    def __init__(self, nc=80, ch=(), drop_rate=0.1, **kwargs):
        super().__init__(nc=nc, ch=ch, **kwargs)
        self.dropout = nn.Dropout2d(p=drop_rate)
    def forward(self, x):
        for i in range(self.nl):
            x[i] = self.dropout(x[i])
        return super().forward(x)

# Register all custom modules in the global and module namespace
globals()['C2f_ECA'] = C2f_ECA
globals()['C2f_CBAM'] = C2f_CBAM
globals()['C2f_SE'] = C2f_SE
globals()['C2f_Attention'] = C2f_Attention
globals()['DropBlock2D'] = DropBlock2D
globals()['Detect_Enhanced'] = Detect_Enhanced
current_module = sys.modules[__name__]
module_classes = {
    'C2f_ECA': C2f_ECA,
    'C2f_CBAM': C2f_CBAM,
    'C2f_SE': C2f_SE,
    'C2f_Attention': C2f_Attention,
    'DropBlock2D': DropBlock2D,
    'Detect_Enhanced': Detect_Enhanced
}
for name, cls in module_classes.items():
    setattr(current_module, name, cls)
print("✅ Custom modules registered: C2f_ECA, C2f_CBAM, C2f_SE, C2f_Attention, DropBlock2D, Detect_Enhanced")
````

## 🔐 4. Advanced Backup & Checkpoint System

````python
# =============================================================================
# COMPREHENSIVE BACKUP & CHECKPOINT SYSTEM
# =============================================================================

class AdvancedCheckpointManager:
    """Advanced checkpoint and backup system for data safety"""
    
    def __init__(self, working_dir, max_backups=5):
        self.working_dir = Path(working_dir)
        self.backup_dir = self.working_dir / "backups"
        self.checkpoint_dir = self.working_dir / "checkpoints"
        self.max_backups = max_backups
        
        # Create directories
        self.backup_dir.mkdir(exist_ok=True)
        self.checkpoint_dir.mkdir(exist_ok=True)
        
        self.session_id = datetime.now().strftime("%Y%m%d_%H%M%S")
        print(f"🔐 Checkpoint Manager initialized - Session: {self.session_id}")
    
    def create_backup(self, stage_name, include_models=True):
        """Create a comprehensive backup of current progress"""
        
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        backup_name = f"{stage_name}_{timestamp}.zip"
        backup_path = self.backup_dir / backup_name
        
        print(f"📦 Creating backup: {backup_name}")
        
        with zipfile.ZipFile(backup_path, 'w', zipfile.ZIP_DEFLATED) as zipf:
            
            # Backup results
            results_dir = self.working_dir / "results"
            if results_dir.exists():
                for file_path in results_dir.rglob('*'):
                    if file_path.is_file():
                        arcname = f"results/{file_path.relative_to(results_dir)}"
                        zipf.write(file_path, arcname)
            
            # Backup logs
            logs_dir = self.working_dir / "logs"
            if logs_dir.exists():
                for file_path in logs_dir.rglob('*'):
                    if file_path.is_file():
                        arcname = f"logs/{file_path.relative_to(logs_dir)}"
                        zipf.write(file_path, arcname)
            
            # Backup checkpoints
            if self.checkpoint_dir.exists():
                for file_path in self.checkpoint_dir.rglob('*'):
                    if file_path.is_file():
                        arcname = f"checkpoints/{file_path.relative_to(self.checkpoint_dir)}"
                        zipf.write(file_path, arcname)
            
            # Backup models (if requested and not too large)
            if include_models:
                models_dir = self.working_dir / "models"
                if models_dir.exists():
                    for file_path in models_dir.rglob('*.pt'):
                        # Only backup small model files (< 100MB)
                        if file_path.stat().st_size < 100 * 1024 * 1024:
                            arcname = f"models/{file_path.relative_to(models_dir)}"
                            zipf.write(file_path, arcname)
            
            # Create metadata
            metadata = {
                'stage': stage_name,
                'timestamp': timestamp,
                'session_id': self.session_id,
                'environment': ENV,
                'device': str(device),
                'python_version': sys.version,
                'torch_version': torch.__version__,
                'system_info': {
                    'cpu_count': psutil.cpu_count(),
                    'memory_gb': psutil.virtual_memory().total / 1024**3,
                    'platform': sys.platform
                }
            }
            
            metadata_str = json.dumps(metadata, indent=2)
            zipf.writestr('metadata.json', metadata_str)
        
        # Clean up old backups
        self._cleanup_old_backups()
        
        backup_size = backup_path.stat().st_size / 1024**2  # MB
        print(f"✅ Backup created: {backup_name} ({backup_size:.1f} MB)")
        
        return backup_path
    
    def save_checkpoint(self, data, checkpoint_name):
        """Save checkpoint data"""
        
        timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")
        checkpoint_file = self.checkpoint_dir / f"{checkpoint_name}_{timestamp}.pkl"
        
        with open(checkpoint_file, 'wb') as f:
            pickle.dump(data, f)
        
        print(f"💾 Checkpoint saved: {checkpoint_file.name}")
        return checkpoint_file
    
    def load_checkpoint(self, checkpoint_pattern):
        """Load the most recent checkpoint matching pattern"""
        
        matching_files = list(self.checkpoint_dir.glob(f"{checkpoint_pattern}*.pkl"))
        if not matching_files:
            print(f"⚠️ No checkpoint found matching: {checkpoint_pattern}")
            return None
        
        # Get most recent
        latest_checkpoint = max(matching_files, key=lambda x: x.stat().st_mtime)
        
        with open(latest_checkpoint, 'rb') as f:
            data = pickle.load(f)
        
        print(f"📂 Checkpoint loaded: {latest_checkpoint.name}")
        return data
    
    def _cleanup_old_backups(self):
        """Keep only the most recent backups"""
        
        backup_files = list(self.backup_dir.glob('*.zip'))
        if len(backup_files) > self.max_backups:
            # Sort by modification time
            backup_files.sort(key=lambda x: x.stat().st_mtime)
            
            # Remove oldest
            for old_backup in backup_files[:-self.max_backups]:
                old_backup.unlink()
                print(f"🗑️ Removed old backup: {old_backup.name}")

# Initialize checkpoint manager
checkpoint_manager = AdvancedCheckpointManager(PATHS['working'])

# Create initial backup
initial_backup = checkpoint_manager.create_backup("00_pipeline_start", include_models=False)
print(f"\n🎯 Pipeline session started with backup: {initial_backup.name}")

# Save initial configuration
initial_config = {
    'environment': ENV,
    'paths': {str(k): str(v) for k, v in PATHS.items()},
    'data_config': DATA_CONFIG if 'DATA_CONFIG' in locals() else None,
    'device': str(device),
    'seed': SEED,
    'timestamp': datetime.now().isoformat()
}

checkpoint_manager.save_checkpoint(initial_config, "initial_config")
print("✅ Initial configuration checkpoint saved!")
````

## 📊 5. Advanced Dataset Analysis & Verification

````python
# =============================================================================
# COMPREHENSIVE DATASET ANALYSIS FOR MEDICAL AI
# =============================================================================

class AdvancedDatasetAnalyzer:
    """Advanced dataset analysis for medical imaging"""
    
    def __init__(self, dataset_path, data_config):
        self.dataset_path = Path(dataset_path)
        self.data_config = data_config
        self.class_names = data_config['names']
        self.nc = data_config['nc']
        
        self.splits = ['train', 'valid', 'test']
        self.analysis_results = {}
        
    def analyze_dataset_structure(self):
        """Analyze and verify dataset structure"""
        
        print("🔍 Analyzing dataset structure...")
        
        structure_status = {}
        total_images = 0
        total_labels = 0
        
        for split in self.splits:
            images_dir = self.dataset_path / split / 'images'
            labels_dir = self.dataset_path / split / 'labels'
            
            # Count files
            image_files = list(images_dir.glob('*.jpg')) + list(images_dir.glob('*.png'))
            label_files = list(labels_dir.glob('*.txt'))
            
            split_images = len(image_files)
            split_labels = len(label_files)
            
            structure_status[split] = {
                'images': split_images,
                'labels': split_labels,
                'match': split_images == split_labels,
                'images_dir_exists': images_dir.exists(),
                'labels_dir_exists': labels_dir.exists()
            }
            
            total_images += split_images
            total_labels += split_labels
            
            print(f"   {split.upper()}: {split_images} images, {split_labels} labels ({'✅' if split_images == split_labels else '❌'})")
        
        print(f"\n📊 Dataset Summary:")
        print(f"   Total Images: {total_images:,}")
        print(f"   Total Labels: {total_labels:,}")
        print(f"   Data Integrity: {'✅ Perfect' if total_images == total_labels else '❌ Mismatch'}")
        
        self.analysis_results['structure'] = structure_status
        self.analysis_results['totals'] = {'images': total_images, 'labels': total_labels}
        
        return structure_status
    
    def analyze_class_distribution(self):
        """Analyze class distribution across splits"""
        
        print("\n🎯 Analyzing class distribution...")
        
        class_counts = {split: Counter() for split in self.splits}
        
        for split in self.splits:
            labels_dir = self.dataset_path / split / 'labels'
            label_files = list(labels_dir.glob('*.txt'))
            
            for label_file in tqdm(label_files, desc=f"Processing {split}", leave=False):
                try:
                    with open(label_file, 'r') as f:
                        lines = f.readlines()
                        for line in lines:
                            parts = line.strip().split()
                            if parts:
                                class_id = int(parts[0])
                                if 0 <= class_id < self.nc:
                                    class_counts[split][class_id] += 1
                except Exception as e:
                    print(f"⚠️ Error reading {label_file}: {e}")
        
        # Create distribution DataFrame
        distribution_data = []
        for class_id in range(self.nc):
            row = {
                'Class_ID': class_id,
                'Class_Name': self.class_names[class_id],
                'Train': class_counts['train'][class_id],
                'Valid': class_counts['valid'][class_id],
                'Test': class_counts['test'][class_id]
            }
            row['Total'] = row['Train'] + row['Valid'] + row['Test']
            distribution_data.append(row)
        
        distribution_df = pd.DataFrame(distribution_data)
        
        # Calculate imbalance metrics
        total_instances = distribution_df['Total'].sum()
        max_instances = distribution_df['Total'].max()
        min_instances = distribution_df['Total'].min()
        imbalance_ratio = max_instances / min_instances if min_instances > 0 else float('inf')
        
        print(f"\n📈 Class Distribution Summary:")
        print(f"   Total Instances: {total_instances:,}")
        print(f"   Most Common: {distribution_df.loc[distribution_df['Total'].idxmax(), 'Class_Name']} ({max_instances} instances)")
        print(f"   Least Common: {distribution_df.loc[distribution_df['Total'].idxmin(), 'Class_Name']} ({min_instances} instances)")
        print(f"   Imbalance Ratio: {imbalance_ratio:.2f}:1")
        
        # Determine imbalance severity
        if imbalance_ratio < 2:
            severity = "Mild"
        elif imbalance_ratio < 5:
            severity = "Moderate"
        elif imbalance_ratio < 10:
            severity = "Severe"
        else:
            severity = "Extreme"
        
        print(f"   Imbalance Severity: {severity}")
        
        self.analysis_results['distribution'] = {
            'dataframe': distribution_df,
            'class_counts': class_counts,
            'imbalance_ratio': imbalance_ratio,
            'severity': severity
        }
        
        return distribution_df
    
    def analyze_image_properties(self, sample_size=100):
        """Analyze image properties (size, quality, etc.)"""
        
        print(f"\n🖼️ Analyzing image properties (sampling {sample_size} images)...")
        
        image_stats = {'widths': [], 'heights': [], 'aspects': [], 'file_sizes': []}
        
        # Sample images from train set
        train_images = list((self.dataset_path / 'train' / 'images').glob('*.jpg'))
        train_images.extend(list((self.dataset_path / 'train' / 'images').glob('*.png')))
        
        if len(train_images) == 0:
            print("⚠️ No training images found to analyze.")
            self.analysis_results['image_properties'] = {}
            return {}
        
        # Set random seed for reproducible sampling
        np.random.seed(42)
        sample_images = np.random.choice(train_images, min(sample_size, len(train_images)), replace=False)
        
        for img_path in tqdm(sample_images, desc="Analyzing images", leave=False):
            try:
                with Image.open(img_path) as img:
                    width, height = img.size
                    # Prevent division by zero and handle invalid aspect ratios
                    if height > 0 and width > 0:
                        aspect_ratio = width / height
                    else:
                        aspect_ratio = 1.0  # Default square aspect ratio
                        print(f"⚠️ Invalid image dimensions: {width}x{height} for {img_path.name}")
                    
                    file_size = img_path.stat().st_size / 1024  # KB
                    
                    image_stats['widths'].append(width)
                    image_stats['heights'].append(height)
                    image_stats['aspects'].append(aspect_ratio)
                    image_stats['file_sizes'].append(file_size)
            except Exception as e:
                print(f"⚠️ Error analyzing {img_path}: {e}")
        
        # Calculate statistics
        stats_summary = {}
        for key, values in image_stats.items():
            if values:
                stats_summary[key] = {
                    'mean': float(np.mean(values)),
                    'std': float(np.std(values)),
                    'min': float(np.min(values)),
                    'max': float(np.max(values)),
                    'median': float(np.median(values))
                }
        
        if stats_summary:
            print(f"\n📏 Image Properties Summary:")
            print(f"   Average Size: {stats_summary['widths']['mean']:.0f} x {stats_summary['heights']['mean']:.0f}")
            print(f"   Size Range: {stats_summary['widths']['min']:.0f}-{stats_summary['widths']['max']:.0f} x {stats_summary['heights']['min']:.0f}-{stats_summary['heights']['max']:.0f}")
            print(f"   Average Aspect Ratio: {stats_summary['aspects']['mean']:.2f}")
            print(f"   Average File Size: {stats_summary['file_sizes']['mean']:.1f} KB")
        
        self.analysis_results['image_properties'] = stats_summary
        
        return stats_summary
    
    def generate_recommendations(self):
        """Generate training recommendations based on analysis"""
        
        print("\n💡 Training Recommendations:")
        
        recommendations = []
        
        # Class imbalance recommendations
        if 'distribution' in self.analysis_results:
            severity = self.analysis_results['distribution']['severity']
            imbalance_ratio = self.analysis_results['distribution']['imbalance_ratio']
            
            if severity in ['Severe', 'Extreme']:
                recommendations.extend([
                    "🎯 Use class-weighted loss functions (inverse frequency weighting)",
                    "🔄 Apply class-specific data augmentation",
                    "📊 Use focal loss to handle hard examples",
                    "⚖️ Consider SMOTE or other synthetic data generation",
                    "📈 Monitor per-class metrics (precision, recall, F1)"
                ])
            
        # Image size recommendations
        if 'image_properties' in self.analysis_results:
            avg_width = self.analysis_results['image_properties']['widths']['mean']
            avg_height = self.analysis_results['image_properties']['heights']['mean']
            
            if avg_width > 1024 or avg_height > 1024:
                recommendations.append("🖼️ Use input size 1024 or higher for medical image quality")
            else:
                recommendations.append("🖼️ Consider input size 640-1024 based on image resolution")
        
        # Medical AI specific recommendations
        recommendations.extend([
            "🏥 Prioritize Recall (Sensitivity) to minimize false negatives",
            "🎚️ Use conservative augmentation to preserve medical features",
            "🔒 Implement deterministic training for reproducibility",
            "📋 Use stratified validation to maintain class ratios",
            "⚡ Apply early stopping with patience for overfitting prevention",
            "🧠 Consider ensemble methods for improved reliability"
        ])
        
        for rec in recommendations:
            print(f"   {rec}")
        
        self.analysis_results['recommendations'] = recommendations
        
        return recommendations

# Initialize and run dataset analysis
print("🚀 Starting comprehensive dataset analysis...")

dataset_analyzer = AdvancedDatasetAnalyzer(PATHS['dataset'], DATA_CONFIG)

# Run all analyses
structure_status = dataset_analyzer.analyze_dataset_structure()
distribution_df = dataset_analyzer.analyze_class_distribution()
image_stats = dataset_analyzer.analyze_image_properties()
recommendations = dataset_analyzer.generate_recommendations()

# Save analysis results
analysis_results = {
    'timestamp': datetime.now().isoformat(),
    'dataset_path': str(PATHS['dataset']),
    'structure_status': structure_status,
    'class_distribution': distribution_df.to_dict('records'),
    'image_properties': image_stats,
    'recommendations': recommendations
}

# Create backup after dataset analysis
checkpoint_manager.save_checkpoint(analysis_results, "dataset_analysis")
backup_path = checkpoint_manager.create_backup("01_dataset_analysis", include_models=False)

print(f"\n✅ Dataset analysis completed and backed up to: {backup_path.name}")
print("🎯 Ready for advanced model architecture definition!")
````

This section implements cutting-edge pre-training techniques specifically designed for medical AI applications:

````python
# =============================================================================
# ADVANCED PRE-TRAINING OPTIMIZATION PIPELINE
# =============================================================================

class AdvancedPreTrainingOptimizer:
    """Advanced pre-training optimizations for medical AI"""
    
    def __init__(self, dataset_path, class_names, analysis_results):
        self.dataset_path = Path(dataset_path)
        self.class_names = class_names
        self.nc = len(class_names)
        self.analysis_results = analysis_results
        
        # Configuration for medical AI
        self.medical_config = {
            'min_confidence': 0.3,      # Lower for medical detection
            'nms_threshold': 0.5,       # Conservative NMS
            'max_det': 300,             # Higher for multiple fractures
            'augment_strength': 0.3,    # Conservative for medical
            'preserve_aspect': True,    # Critical for medical imaging
        }
    
    def create_smart_augmentation_pipeline(self):
        """Create medical-specific augmentation pipeline"""
        
        print("🎨 Creating smart augmentation pipeline...")
        
        # Get image statistics for adaptive augmentation
        img_stats = self.analysis_results.get('image_properties', {})
        
        # Medical-safe augmentations
        augmentations = {
            # Spatial augmentations (preserve anatomy)
            'hsv_h': 0.015,      # Minimal hue shift
            'hsv_s': 0.3,        # Moderate saturation
            'hsv_v': 0.2,        # Brightness variation
            'degrees': 5,        # Small rotation only
            'translate': 0.05,   # Minimal translation
            'scale': 0.1,        # Small scale changes
            'shear': 2,          # Minimal shear
            'perspective': 0.0001,# Tiny perspective change
            'flipud': 0.0,       # No vertical flip (anatomically incorrect)
            'fliplr': 0.5,       # Horizontal flip OK
            'mosaic': 0.8,       # Mosaic for context
            'mixup': 0.1,        # Minimal mixup
            'copy_paste': 0.1,   # Conservative copy-paste
            
            # Medical-specific enhancements
            'sharpen': 0.2,      # Enhance bone definition
            'noise': 0.1,        # Simulate imaging noise
            'blur': 0.05,        # Minimal blur simulation
        }
        
        print("   📋 Medical-Safe Augmentation Parameters:")
        for key, value in augmentations.items():
            print(f"      {key}: {value}")
        
        return augmentations
    
    def implement_intelligent_class_balancing(self):
        """Implement intelligent class balancing strategy"""
        
        print("\n⚖️ Implementing intelligent class balancing...")
        
        if 'distribution' not in self.analysis_results:
            print("   ⚠️ No distribution analysis available. Using default weights.")
            return None
        
        class_counts = self.analysis_results['distribution']['class_counts']['train']
        total_samples = sum(class_counts.values())
        
        # Calculate inverse frequency weights with division by zero protection
        class_weights = {}
        
        # Validate inputs
        if total_samples == 0:
            print("   ⚠️ No training samples found. Using equal weights.")
            return {i: 1.0 for i in range(self.nc)}
            
        if self.nc == 0:
            print("   ⚠️ No classes defined. Cannot calculate weights.")
            return None
        
        for class_id in range(self.nc):
            count = class_counts.get(class_id, 0)
            # Prevent division by zero
            if count > 0:
                weight = total_samples / (self.nc * count)
            else:
                weight = 1.0  # Default weight for missing classes
                print(f"   ⚠️ Class {class_id} has no samples, using default weight")
            class_weights[class_id] = weight
        
        # Normalize weights to prevent extreme values
        if class_weights:
            max_weight = max(class_weights.values())
            min_weight = min(class_weights.values())
            
            # Cap maximum weight ratio for stability (prevent division by zero)
            if min_weight > 0 and max_weight / min_weight > 10:
                cap_ratio = 10
                for class_id in class_weights:
                    if class_weights[class_id] > min_weight * cap_ratio:
                        class_weights[class_id] = min_weight * cap_ratio
        
        print("   📊 Class Weights:")
        for class_id, weight in class_weights.items():
            class_name = self.class_names[class_id]
            count = class_counts.get(class_id, 0)
            print(f"      {class_name} (ID {class_id}): {weight:.3f} (samples: {count})")
        
        return class_weights
    
    def setup_progressive_training_strategy(self):
        """Setup progressive/curriculum learning strategy"""
        
        print("\n📚 Setting up progressive training strategy...")
        
        # Progressive training phases
        training_phases = [
            {
                'name': 'Foundation Phase',
                'epochs': 50,
                'img_size': 320,
                'batch_size': 32,
                'lr0': 0.01,
                'augment_strength': 0.1,
                'focus': 'Basic feature learning'
            },
            {
                'name': 'Development Phase', 
                'epochs': 100,
                'img_size': 640,
                'batch_size': 16,
                'lr0': 0.005,
                'augment_strength': 0.3,
                'focus': 'Pattern recognition'
            },
            {
                'name': 'Refinement Phase',
                'epochs': 100,
                'img_size': 1024,
                'batch_size': 8,
                'lr0': 0.001,
                'augment_strength': 0.5,
                'focus': 'Fine-grained detection'
            }
        ]
        
        print("   🎯 Training Phase Strategy:")
        for i, phase in enumerate(training_phases, 1):
            print(f"      Phase {i}: {phase['name']}")
            print(f"         Epochs: {phase['epochs']}")
            print(f"         Image Size: {phase['img_size']}")
            print(f"         Batch Size: {phase['batch_size']}")
            print(f"         Learning Rate: {phase['lr0']}")
            print(f"         Focus: {phase['focus']}")
            print()
        
        return training_phases
    
    def create_quality_aware_data_selection(self):
        """Create quality-aware data selection pipeline"""
        
        print("🔍 Creating quality-aware data selection...")
        
        quality_criteria = {
            'min_image_size': (224, 224),
            'max_aspect_ratio': 3.0,
            'min_aspect_ratio': 0.33,
            'min_file_size_kb': 10,
            'max_file_size_mb': 50,
            'required_channels': 3,
            'min_brightness': 20,
            'max_brightness': 235
        }
        
        print("   📋 Quality Criteria:")
        for criterion, value in quality_criteria.items():
            print(f"      {criterion}: {value}")
        
        return quality_criteria
    
    def setup_stratified_cross_validation(self, n_splits=5):
        """Setup stratified k-fold cross-validation"""
        
        print(f"\n🔄 Setting up {n_splits}-fold stratified cross-validation...")
        
        cv_config = {
            'n_splits': n_splits,
            'shuffle': True,
            'random_state': 42,
            'stratify_by': 'class_distribution',
            'validation_metrics': [
                'precision', 'recall', 'f1-score', 
                'ap_50', 'ap_50_95', 'confusion_matrix'
            ]
        }
        
        print("   📊 Cross-Validation Configuration:")
        for key, value in cv_config.items():
            print(f"      {key}: {value}")
        
        return cv_config
    
    def generate_optimized_training_config(self):
        """Generate optimized training configuration"""
        
        print("\n⚙️ Generating optimized training configuration...")
        
        # Get class weights
        class_weights = self.implement_intelligent_class_balancing()
        
        # Get augmentation pipeline
        augmentations = self.create_smart_augmentation_pipeline()
        
        # Get progressive strategy
        training_phases = self.setup_progressive_training_strategy()
        
        # Get quality criteria
        quality_criteria = self.create_quality_aware_data_selection()
        
        # Get CV configuration
        cv_config = self.setup_stratified_cross_validation()
        
        # Optimized training configuration
        optimized_config = {
            'model_config': {
                'model_size': 'yolov10n',
                'input_size': 640,  # Start with medium size
                'pretrained': True,
                'freeze_backbone': 10,  # Freeze first 10 layers initially
            },
            
            'training_config': {
                'epochs': 250,
                'batch_size': 16,
                'patience': 50,
                'save_period': 10,
                'cache': True,
                'device': 'auto',
                'workers': 8,
                'exist_ok': True,
                'seed': 42,
                
                # Optimizer settings
                'optimizer': 'AdamW',
                'lr0': 0.002,
                'lrf': 0.1,
                'momentum': 0.9,
                'weight_decay': 0.0005,
                'warmup_epochs': 5,
                'warmup_momentum': 0.8,
                'warmup_bias_lr': 0.1,
                
                # Loss function
                'box': 7.5,      # Box loss gain
                'cls': 0.5,      # Class loss gain  
                'dfl': 1.5,      # DFL loss gain
                'label_smoothing': 0.1,
                
                # Medical AI specific
                'conf': 0.25,    # Confidence threshold
                'iou': 0.7,      # IoU threshold for NMS
                'max_det': 300,  # Maximum detections
                'half': True,    # Use half precision
                'dnn': False,    # Use OpenCV DNN
                'plots': True,   # Save plots
                'val': True,     # Validate during training
            },
            
            'augmentation_config': augmentations,
            'class_weights': class_weights,
            'training_phases': training_phases,
            'quality_criteria': quality_criteria,
            'cv_config': cv_config,
            'medical_config': self.medical_config
        }
        
        print("   ✅ Optimized configuration generated!")
        
        return optimized_config

# Initialize pre-training optimizer
print("🚀 Initializing advanced pre-training optimizer...")

pre_training_optimizer = AdvancedPreTrainingOptimizer(
    PATHS['dataset'], 
    DATA_CONFIG['names'], 
    dataset_analyzer.analysis_results
)

# Generate optimized configuration
optimized_training_config = pre_training_optimizer.generate_optimized_training_config()

# Save optimized configuration
config_path = PATHS['working'] / 'optimized_training_config.yaml'
with open(config_path, 'w') as f:
    yaml.dump(optimized_training_config, f, default_flow_style=False, sort_keys=False)

print(f"\n✅ Pre-training optimizations completed!")
print(f"📄 Configuration saved to: {config_path}")
print("🎯 Ready for custom model architecture definition!")
````

````python
# =============================================================================
# ADVANCED ATTENTION MECHANISMS FOR MEDICAL IMAGING
# =============================================================================

class ECAAttention(nn.Module):
    """Efficient Channel Attention (ECA) - Lightweight and effective"""
    
    def __init__(self, channels, gamma=2, b=1):
        super(ECAAttention, self).__init__()
        
        # Validate inputs to prevent mathematical errors
        if channels <= 0:
            raise ValueError(f"Number of channels must be positive, got {channels}")
        if gamma <= 0:
            raise ValueError(f"Gamma must be positive, got {gamma}")
        
        # Calculate kernel size adaptively with proper domain validation
        try:
            # Ensure channels > 0 for log2
            log_val = np.log2(max(channels, 1))  # Prevent log2(0)
            t = int(abs((log_val + b) / gamma))
            k = max(t if t % 2 else t + 1, 3)  # Ensure minimum kernel size of 3
        except (ValueError, ZeroDivisionError) as e:
            print(f"⚠️ Warning in ECA kernel calculation: {e}, using default kernel size 3")
            k = 3
        
        self.avg_pool = nn.AdaptiveAvgPool2d(1)
        self.conv = nn.Conv1d(1, 1, kernel_size=k, padding=k//2, bias=False)
        self.sigmoid = nn.Sigmoid()
    
    def forward(self, x):
        # Global average pooling
        y = self.avg_pool(x)
        
        # 1D convolution along channel dimension
        y = self.conv(y.squeeze(-1).transpose(-1, -2))
        y = y.transpose(-1, -2).unsqueeze(-1)
        
        # Apply attention
        y = self.sigmoid(y)
        return x * y.expand_as(x)

class CBAM(nn.Module):
    """Convolutional Block Attention Module - Channel + Spatial"""
    
    def __init__(self, channels, reduction=16, kernel_size=7):
        super(CBAM, self).__init__()
        
        # Validate inputs to prevent division errors
        if channels <= 0:
            raise ValueError(f"Number of channels must be positive, got {channels}")
        if reduction <= 0:
            raise ValueError(f"Reduction factor must be positive, got {reduction}")
        
        # Ensure reduction doesn't make channels too small
        reduced_channels = max(channels // reduction, 1)  # Prevent 0 channels
        
        # Channel attention
        self.channel_attention = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Conv2d(channels, reduced_channels, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(reduced_channels, channels, 1, bias=False)
        )
        
        self.channel_attention_max = nn.Sequential(
            nn.AdaptiveMaxPool2d(1),
            nn.Conv2d(channels, reduced_channels, 1, bias=False),
            nn.ReLU(inplace=True),
            nn.Conv2d(reduced_channels, channels, 1, bias=False)
        )
        
        # Spatial attention
        self.spatial_attention = nn.Sequential(
            nn.Conv2d(2, 1, kernel_size, padding=kernel_size//2, bias=False),
            nn.Sigmoid()
        )
        
        self.sigmoid = nn.Sigmoid()
    
    def forward(self, x):
        # Channel attention
        avg_out = self.channel_attention(x)
        max_out = self.channel_attention_max(x)
        channel_att = self.sigmoid(avg_out + max_out)
        x = x * channel_att
        
        # Spatial attention
        avg_out = torch.mean(x, dim=1, keepdim=True)
        max_out, _ = torch.max(x, dim=1, keepdim=True)
        spatial_input = torch.cat([avg_out, max_out], dim=1)
        spatial_att = self.spatial_attention(spatial_input)
        x = x * spatial_att
        
        return x

class SEBlock(nn.Module):
    """Squeeze-and-Excitation Block - Global context modeling"""
    
    def __init__(self, channels, reduction=16):
        super(SEBlock, self).__init__()
        
        # Validate inputs to prevent division errors
        if channels <= 0:
            raise ValueError(f"Number of channels must be positive, got {channels}")
        if reduction <= 0:
            raise ValueError(f"Reduction factor must be positive, got {reduction}")
        
        # Ensure reduction doesn't make channels too small
        reduced_channels = max(channels // reduction, 1)  # Prevent 0 channels
        
        self.global_avgpool = nn.AdaptiveAvgPool2d(1)
        self.fc = nn.Sequential(
            nn.Linear(channels, reduced_channels, bias=False),
            nn.ReLU(inplace=True),
            nn.Linear(reduced_channels, channels, bias=False),
            nn.Sigmoid()
        )
    
    def forward(self, x):
        b, c, _, _ = x.size()
        y = self.global_avgpool(x).view(b, c)
        y = self.fc(y).view(b, c, 1, 1)
        return x * y.expand_as(x)

class DropBlock2D(nn.Module):
    """DropBlock regularization for spatial features"""
    
    def __init__(self, drop_rate=0.1, block_size=7):
        super(DropBlock2D, self).__init__()
        # Validate inputs to prevent mathematical errors
        if not (0.0 <= drop_rate <= 1.0):
            raise ValueError(f"Drop rate must be between 0.0 and 1.0, got {drop_rate}")
        if block_size <= 0:
            raise ValueError(f"Block size must be positive, got {block_size}")
        
        self.drop_rate = drop_rate
        self.block_size = max(block_size, 1)  # Ensure minimum block size
    
    def forward(self, x):
        if not self.training or self.drop_rate == 0.0:
            return x
        
        # Calculate gamma with validation
        try:
            gamma = self.drop_rate / (self.block_size ** 2)
        except ZeroDivisionError:
            print("⚠️ Warning: Block size is 0, skipping DropBlock")
            return x
        
        # Sample mask
        mask = (torch.rand(x.shape[0], *x.shape[2:]) < gamma).float()
        
        # Apply block mask
        mask = mask.to(x.device)
        block_mask = self._compute_block_mask(mask)
        
        # Normalize and apply with division by zero protection
        countM = block_mask.size()[0] * block_mask.size()[1]
        count_ones = block_mask.sum()
        
        if count_ones > 0:
            return x * block_mask.unsqueeze(1) * (countM / count_ones)
        else:
            # If no elements to drop, return original input
            return x
    
    def _compute_block_mask(self, mask):
        block_mask = F.max_pool2d(
            input=mask.unsqueeze(1),
            kernel_size=(self.block_size, self.block_size),
            stride=(1, 1),
            padding=self.block_size // 2
        )
        
        if self.block_size % 2 == 0:
            block_mask = block_mask[:, :, :-1, :-1]
        
        block_mask = 1 - block_mask.squeeze(1)
        return block_mask

class FocalLoss(nn.Module):
    """Focal Loss for addressing class imbalance"""
    
    def __init__(self, alpha=1, gamma=2, reduction='mean'):
        super(FocalLoss, self).__init__()
        self.alpha = alpha
        self.gamma = gamma
        self.reduction = reduction
    
    def forward(self, inputs, targets):
        ce_loss = F.cross_entropy(inputs, targets, reduction='none')
        pt = torch.exp(-ce_loss)
        focal_loss = self.alpha * (1-pt)**self.gamma * ce_loss
        
        if self.reduction == 'mean':
            return focal_loss.mean()
        elif self.reduction == 'sum':
            return focal_loss.sum()
        else:
            return focal_loss

print("✅ Advanced attention mechanisms defined!")
print("   🔍 ECA: Efficient Channel Attention")
print("   🎯 CBAM: Channel + Spatial Attention") 
print("   🌐 SE: Squeeze-and-Excitation")
print("   🎲 DropBlock: Spatial regularization")
print("   🎯 Focal Loss: Class imbalance handling")

# Use FocalLoss as the loss function in your training loop
criterion = FocalLoss(alpha=1, gamma=2)
# Example:
## for images, targets in dataloader:
##     outputs = model(images)
##     loss = criterion(outputs, targets)
##     loss.backward()
##     optimizer.step()
````

````python

# ADVANCED MODEL ARCHITECTURE & YAML GENERATION (LOCAL MACHINE)


class AdvancedYOLOv10nBuilder:
    def __init__(self, num_classes, input_size=1024):
        self.num_classes = num_classes
        self.input_size = input_size
        self.config = self._create_config()

    def _c2f_args(self, c1, c2=None, n=3):
        if c2 is None:
            c2 = c1
        return [int(c1), int(c2), int(n)]

    def _dropblock_args(self, drop_rate=0.1, block_size=7):
        return [float(drop_rate), int(block_size)]

    def _upsample_args(self, scale=2):
        return [scale]

    def _layer(self, from_idx, num, module, args):
        # Only Concat and Detect_Enhanced can have list input indices
        if module in ['Concat', 'Detect_Enhanced']:
            return [from_idx, num, module, args]
        # All other layers must have a single integer as input index
        if isinstance(from_idx, list):
            if len(from_idx) == 1:
                from_idx = from_idx[0]
            else:
                # Fallback: use last index (should not happen, but safe)
                from_idx = from_idx[-1]
        return [from_idx, num, module, args]

    def _create_config(self):
        outputs = []
        backbone = []
        head = []
        # Backbone layers
        for spec in [
            ('Conv', [64, 3, 2]),
            ('Conv', [128, 3, 2]),
            ('C2f_ECA', self._c2f_args(128, 128, 3)),
            ('C2f_CBAM', self._c2f_args(128, 128, 3)),
            ('C2f_SE', self._c2f_args(128, 128, 3)),
            ('C2f_Attention', self._c2f_args(128, 128, 3)),
            ('DropBlock2D', self._dropblock_args(0.1, 7)),
            ('Conv', [256, 3, 2]),
            ('C2f_ECA', self._c2f_args(256, 256, 3)),
            ('C2f_CBAM', self._c2f_args(256, 256, 3)),
            ('C2f_SE', self._c2f_args(256, 256, 3)),
            ('C2f_Attention', self._c2f_args(256, 256, 3)),
            ('DropBlock2D', self._dropblock_args(0.1, 7)),
            ('Conv', [512, 3, 2]),
            ('C2f_ECA', self._c2f_args(512, 512, 3)),
            ('C2f_CBAM', self._c2f_args(512, 512, 3)),
            ('C2f_SE', self._c2f_args(512, 512, 3)),
            ('C2f_Attention', self._c2f_args(512, 512, 3)),
            ('DropBlock2D', self._dropblock_args(0.1, 7)),
            ('Conv', [1024, 3, 2]),
            ('SPPF', [1024, 5]),
        ]:
            idx = outputs[-1] if outputs else -1
            backbone.append(self._layer(idx, 1, spec[0], spec[1]))
            outputs.append(len(backbone)-1 if outputs else 0)
        # Head layers (track outputs for correct Concat indices)
        featuremap_indices = []
        idx_16 = 16  # C2f_CBAM (512, 512, 3)
        idx_10 = 10  # C2f_CBAM (256, 256, 3)
        idx_19 = 19  # DropBlock2D (512, 512, 3)
        idx_21 = 21  # SPPF
        head_idx = len(backbone) - 1
        # 1st Upsample + Concat
        head.append(self._layer(head_idx, 1, 'nn.Upsample', self._upsample_args(2)))
        head_idx += 1
        head.append(self._layer([head_idx, idx_16], 1, 'Concat', [1]))
        head_idx += 1
        head.append(self._layer(head_idx, 3, 'C2f_ECA', self._c2f_args(512, 512, 3)))
        featuremap_indices.append(head_idx)
        head_idx += 1
        # 2nd Upsample + Concat
        head.append(self._layer(head_idx, 1, 'nn.Upsample', self._upsample_args(2)))
        head_idx += 1
        head.append(self._layer([head_idx, idx_10], 1, 'Concat', [1]))
        head_idx += 1
        head.append(self._layer(head_idx, 3, 'C2f_CBAM', self._c2f_args(256, 256, 3)))
        featuremap_indices.append(head_idx)
        head_idx += 1
        # Downsample + Concat
        head.append(self._layer(head_idx, 1, 'Conv', [256, 3, 2]))
        head_idx += 1
        head.append(self._layer([head_idx, idx_19], 1, 'Concat', [1]))
        head_idx += 1
        head.append(self._layer(head_idx, 3, 'C2f_SE', self._c2f_args(512, 512, 3)))
        featuremap_indices.append(head_idx)
        head_idx += 1
        # Downsample + Concat
        head.append(self._layer(head_idx, 1, 'Conv', [512, 3, 2]))
        head_idx += 1
        head.append(self._layer([head_idx, idx_21], 1, 'Concat', [1]))
        head_idx += 1
        head.append(self._layer(head_idx, 3, 'C2f_Attention', self._c2f_args(1024, 1024, 3)))
        head_idx += 1
        # Detect_Enhanced uses the last three featuremap indices
        detect_indices = featuremap_indices[-3:]
        head.append(self._layer(detect_indices, 1, 'Detect_Enhanced', [self.num_classes, 0.1]))
        config = {
            'nc': self.num_classes,
            'depth_multiple': 0.33,
            'width_multiple': 0.25,
            'max_channels': 1024,
            'backbone': backbone,
            'head': head
        }
        return config

    def create_enhanced_yaml(self, save_path):
        import yaml
        save_path = Path(save_path)
        architecture_config = {
            'nc': self.config['nc'],
            'depth_multiple': self.config['depth_multiple'],
            'width_multiple': self.config['width_multiple'],
            'max_channels': self.config['max_channels'],
            'backbone': self.config['backbone'],
            'head': self.config['head']
        }
        save_path.parent.mkdir(parents=True, exist_ok=True)
        with open(save_path, 'w', encoding='utf-8') as f:
            yaml.dump(architecture_config, f, default_flow_style=False, sort_keys=False, indent=2)
        print(f"✅ Enhanced YOLOv10n config saved to: {save_path}")
        return save_path

# Create and save enhanced model configuration
model_builder = AdvancedYOLOv10nBuilder(num_classes=DATA_CONFIG['nc'], input_size=1024)
config_path = PATHS['working'] / 'models' / 'enhanced_yolov10n.yaml'
enhanced_config_path = model_builder.create_enhanced_yaml(config_path)
print(f"\n🧠 Enhanced YOLOv10n Architecture Summary:")
print(f"   📊 Classes: {DATA_CONFIG['nc']}")
print(f"   🖼️ Input Size: 1024x1024")
print(f"   🎯 Architecture: All custom modules and advanced regularization included")
print(f"   📁 Config: {enhanced_config_path}")
````

````python
# =============================================================================
# MEDICAL-SAFE ADVANCED AUGMENTATION PIPELINE
# =============================================================================

class MedicalAugmentationPipeline:
    """Advanced augmentation pipeline optimized for medical imaging"""
    
    def __init__(self, input_size=1024, severity='moderate'):
        self.input_size = input_size
        self.severity = severity
        self.augmentation_configs = self._create_augmentation_configs()
    
    def _create_augmentation_configs(self):
        """Create different augmentation configurations based on severity"""
        
        configs = {}
        
        # Conservative augmentation for medical data
        configs['conservative'] = {
            'geometric': {
                'rotation_limit': 5,
                'shift_limit': 0.05,
                'scale_limit': 0.1,
                'shear_limit': 3,
                'perspective_limit': 0.02
            },
            'photometric': {
                'brightness_limit': 0.1,
                'contrast_limit': 0.1,
                'saturation_limit': 0.1,
                'hue_limit': 5,
                'gamma_limit': (90, 110)
            },
            'noise': {
                'gaussian_noise_var': (5, 15),
                'blur_limit': 3,
                'motion_blur_limit': 3
            }
        }
        
        # Moderate augmentation 
        configs['moderate'] = {
            'geometric': {
                'rotation_limit': 10,
                'shift_limit': 0.1,
                'scale_limit': 0.2,
                'shear_limit': 5,
                'perspective_limit': 0.05
            },
            'photometric': {
                'brightness_limit': 0.2,
                'contrast_limit': 0.2,
                'saturation_limit': 0.2,
                'hue_limit': 10,
                'gamma_limit': (80, 120)
            },
            'noise': {
                'gaussian_noise_var': (10, 25),
                'blur_limit': 5,
                'motion_blur_limit': 5
            }
        }
        
        # Aggressive augmentation for robust training
        configs['aggressive'] = {
            'geometric': {
                'rotation_limit': 15,
                'shift_limit': 0.15,
                'scale_limit': 0.3,
                'shear_limit': 8,
                'perspective_limit': 0.1
            },
            'photometric': {
                'brightness_limit': 0.3,
                'contrast_limit': 0.3,
                'saturation_limit': 0.3,
                'hue_limit': 15,
                'gamma_limit': (70, 130)
            },
            'noise': {
                'gaussian_noise_var': (15, 35),
                'blur_limit': 7,
                'motion_blur_limit': 7
            }
        }
        
        return configs
    
    def create_training_augmentation(self):
        """Create training augmentation pipeline"""
        
        config = self.augmentation_configs[self.severity]
        
        augmentation = A.Compose([
            # Geometric transformations (medical-safe)
            A.Rotate(
                limit=config['geometric']['rotation_limit'],
                interpolation=cv2.INTER_LINEAR,
                border_mode=cv2.BORDER_REFLECT_101,
                p=0.5
            ),
            
            A.ShiftScaleRotate(
                shift_limit=config['geometric']['shift_limit'],
                scale_limit=config['geometric']['scale_limit'],
                rotate_limit=config['geometric']['rotation_limit'],
                interpolation=cv2.INTER_LINEAR,
                border_mode=cv2.BORDER_REFLECT_101,
                p=0.6
            ),
            
            A.Perspective(
                scale=config['geometric']['perspective_limit'],
                p=0.3
            ),
            
            # Photometric transformations
            A.RandomBrightnessContrast(
                brightness_limit=config['photometric']['brightness_limit'],
                contrast_limit=config['photometric']['contrast_limit'],
                p=0.5
            ),
            
            A.HueSaturationValue(
                hue_shift_limit=config['photometric']['hue_limit'],
                sat_shift_limit=config['photometric']['saturation_limit'],
                val_shift_limit=config['photometric']['brightness_limit'],
                p=0.4
            ),
            
            A.RandomGamma(
                gamma_limit=config['photometric']['gamma_limit'],
                p=0.3
            ),
            
            # Noise and blur (conservative for medical)
            A.OneOf([
                A.GaussNoise(
                    var_limit=config['noise']['gaussian_noise_var'],
                    p=0.3
                ),
                A.Blur(
                    blur_limit=config['noise']['blur_limit'],
                    p=0.2
                ),
                A.MotionBlur(
                    blur_limit=config['noise']['motion_blur_limit'],
                    p=0.2
                ),
            ], p=0.4),
            
            # Medical-specific augmentations
            A.CLAHE(
                clip_limit=2.0,
                tile_grid_size=(8, 8),
                p=0.3
            ),
            
            A.Sharpen(
                alpha=(0.1, 0.3),
                lightness=(0.8, 1.2),
                p=0.2
            ),
            
            # Cutout variants (spatial regularization)
            A.OneOf([
                A.CoarseDropout(
                    max_holes=8,
                    max_height=32,
                    max_width=32,
                    min_holes=1,
                    min_height=8,
                    min_width=8,
                    fill_value=0,
                    p=0.3
                ),
                A.GridDropout(
                    ratio=0.3,
                    unit_size_min=8,
                    unit_size_max=32,
                    holes_number_x=4,
                    holes_number_y=4,
                    p=0.2
                ),
            ], p=0.3),
            
            # Final resize and normalization
            A.Resize(
                height=self.input_size,
                width=self.input_size,
                interpolation=cv2.INTER_LINEAR,
                p=1.0
            ),
            
            A.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225],
                p=1.0
            ),
            
            ToTensorV2(p=1.0)
        ], bbox_params=A.BboxParams(
            format='yolo',
            label_fields=['class_labels'],
            min_visibility=0.3
        ))
        
        return augmentation
    
    def create_validation_augmentation(self):
        """Create validation augmentation pipeline (minimal)"""
        
        augmentation = A.Compose([
            A.Resize(
                height=self.input_size,
                width=self.input_size,
                interpolation=cv2.INTER_LINEAR,
                p=1.0
            ),
            
            A.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225],
                p=1.0
            ),
            
            ToTensorV2(p=1.0)
        ], bbox_params=A.BboxParams(
            format='yolo',
            label_fields=['class_labels']
        ))
        
        return augmentation
    
    def create_test_time_augmentation(self):
        """Create test-time augmentation for ensemble predictions"""
        
        tta_transforms = []
        
        # Original
        tta_transforms.append(A.Compose([
            A.Resize(self.input_size, self.input_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ]))
        
        # Horizontal flip
        tta_transforms.append(A.Compose([
            A.HorizontalFlip(p=1.0),
            A.Resize(self.input_size, self.input_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ]))
        
        # Slight rotation
        tta_transforms.append(A.Compose([
            A.Rotate(limit=5, p=1.0),
            A.Resize(self.input_size, self.input_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ]))
        
        # Brightness adjustment
        tta_transforms.append(A.Compose([
            A.RandomBrightnessContrast(brightness_limit=0.1, contrast_limit=0.1, p=1.0),
            A.Resize(self.input_size, self.input_size),
            A.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225]),
            ToTensorV2()
        ]))
        
        return tta_transforms

# Initialize augmentation pipeline
print("🎨 Initializing advanced medical augmentation pipeline...")

aug_pipeline = MedicalAugmentationPipeline(
    input_size=1024,
    severity='moderate'  # Balanced for medical imaging
)

# Create augmentation transforms
train_transform = aug_pipeline.create_training_augmentation()
val_transform = aug_pipeline.create_validation_augmentation()
tta_transforms = aug_pipeline.create_test_time_augmentation()

print(f"✅ Augmentation pipeline created:")
print(f"   🔄 Training: Medical-safe moderate augmentation")
print(f"   ✅ Validation: Minimal preprocessing only")
print(f"   🎯 TTA: {len(tta_transforms)} test-time augmentations")
print(f"   🖼️ Input size: {aug_pipeline.input_size}x{aug_pipeline.input_size}")

# Save augmentation configuration
aug_config = {
    'severity': aug_pipeline.severity,
    'input_size': aug_pipeline.input_size,
    'configs': aug_pipeline.augmentation_configs,
    'tta_count': len(tta_transforms)
}

checkpoint_manager.save_checkpoint(aug_config, "augmentation_config")
print("💾 Augmentation configuration saved to checkpoint!")
````

````python
# =============================================================================
# ADVANCED TRAINING MANAGER WITH ERROR HANDLING
# =============================================================================

class SimpleCheckpointManager:
    """Simple checkpoint manager for basic functionality"""
    
    def __init__(self):
        self.checkpoint_dir = PATHS['working'] / 'checkpoints'
        self.checkpoint_dir.mkdir(exist_ok=True)
    
    def optimize_training_config(self, config):
        """Simple config optimization - just return the original config"""
        return config
    
    def save_checkpoint(self, model, epoch, metrics):
        """Save a checkpoint"""
        try:
            checkpoint_path = self.checkpoint_dir / f'checkpoint_epoch_{epoch}.pt'
            checkpoint = {
                'epoch': epoch,
                'model_state_dict': model.state_dict() if hasattr(model, 'state_dict') else None,
                'metrics': metrics
            }
            torch.save(checkpoint, checkpoint_path)
            return checkpoint_path
        except Exception as e:
            print(f"⚠️ Failed to save checkpoint: {e}")
            return None

class AdvancedTrainingManager:
    """Advanced training manager with comprehensive error handling"""
    
    def __init__(self, data_config, model_builder, checkpoint_manager=None, overfitting_validator=None):
        self.data_config = data_config
        self.model_builder = model_builder
        self.checkpoint_manager = checkpoint_manager or SimpleCheckpointManager()
        self.overfitting_validator = overfitting_validator
        
        # Create directories
        self.models_dir = PATHS['working'] / 'models'
        self.results_dir = PATHS['working'] / 'training_results'
        self.models_dir.mkdir(exist_ok=True)
        self.results_dir.mkdir(exist_ok=True)
        
        # Available training strategies
        self.strategies = {
            'precision_optimized_v1': {
                'description': 'Targets realistic medical precision (85-90%)',
                'epochs': 150,
                'batch': 8,
                'lr0': 0.0005,
                'weight_decay': 0.001,
                'label_smoothing': 0.2,
                'cls': 1.5,  # Higher classification loss weight
                'box': 7.5,  # Standard box regression weight
                'target_precision': 0.95,
                'target_recall': 0.90
            },
            'recall_optimized_v1': {
                'description': 'Minimizes false negatives',
                'epochs': 200,
                'batch': 6,
                'lr0': 0.0003,
                'weight_decay': 0.0005,
                'label_smoothing': 0.1,
                'cls': 2.0,  # Higher classification weight
                'box': 8.0,  # Higher box weight for better localization
                'target_precision': 0.80,
                'target_recall': 0.95
            },
            'balanced_v1': {
                'description': 'Optimizes overall F1-score',
                'epochs': 175,
                'batch': 8,
                'lr0': 0.0004,
                'weight_decay': 0.0008,
                'label_smoothing': 0.15,
                'cls': 1.0,
                'box': 7.5,
                'target_precision': 0.88,
                'target_recall': 0.88
            }
        }
    
    def train_model(self, strategy='precision_optimized_v1', experiment_name=None, resume_from=None):
        """Train model with comprehensive error handling and fallback strategies"""
        
        if strategy not in self.strategies:
            raise ValueError(f"Unknown strategy: {strategy}")
        
        if experiment_name is None:
            experiment_name = f"{strategy}_{datetime.now().strftime('%Y%m%d_%H%M%S')}"
        
        strategy_config = self.strategies[strategy].copy()
        print(f"🚀 Starting advanced training: {experiment_name}")
        print(f"📊 Strategy: {strategy_config['description']}")
        print(f"🎯 Target: >{strategy_config['target_precision']*100:.0f}% precision, >{strategy_config['target_recall']*100:.0f}% recall")
        print("="*60)
        
        try:
            # Create experiment directory
            experiment_dir = self.results_dir / experiment_name
            experiment_dir.mkdir(exist_ok=True)
            
            # Build training configuration
            base_config = {
                'data': str(PATHS['data_yaml']),
                'epochs': strategy_config['epochs'],
                'batch': strategy_config['batch'],
                'imgsz': 640,
                'device': str(device).replace('cuda:', ''),
                'workers': 4,
                'project': str(experiment_dir.parent),
                'name': experiment_name,
                'exist_ok': True,
                'save': True,
                'save_period': 10,
                'cache': False,  # Disable cache for Kaggle
                'verbose': True,
                'seed': SEED,
                'deterministic': True,
                'single_cls': False,
                'rect': False,
                'cos_lr': True,
                'close_mosaic': 10,
                'resume': resume_from is not None,
                'amp': True,  # Automatic Mixed Precision
                'fraction': 1.0,
                'profile': False,
                'freeze': None,
                'lr0': strategy_config['lr0'],
                'lrf': 0.01,
                'momentum': 0.937,
                'weight_decay': strategy_config['weight_decay'],
                'warmup_epochs': 3,
                'warmup_momentum': 0.8,
                'warmup_bias_lr': 0.1,
                'box': strategy_config['box'],
                'cls': strategy_config['cls'],
                'dfl': 1.5,
                'pose': 12.0,
                'kobj': 1.0,
                'label_smoothing': strategy_config['label_smoothing'],
                'nbs': 64,
                'overlap_mask': True,
                'mask_ratio': 4,
                'dropout': 0.0,
                'val': True,
                'split': 'val',
                'save_json': False,
                'save_hybrid': False,
                'conf': None,
                'iou': 0.7,
                'max_det': 300,
                'half': False,
                'dnn': False,
                'plots': True,
            }
            
            # Apply checkpoint manager optimizations
            try:
                optimized_config = self.checkpoint_manager.optimize_training_config(base_config)
            except Exception as e:
                print(f"⚠️ Checkpoint manager optimization failed: {e}")
                optimized_config = base_config
            
            # Get validation configuration from overfitting validator
            try:
                if self.overfitting_validator:
                    validation_config = self.overfitting_validator.get_validation_config()
                    final_config = {**optimized_config, **validation_config}
                else:
                    final_config = optimized_config
            except Exception as e:
                print(f"⚠️ Overfitting validator config failed: {e}")
                final_config = optimized_config
            
            print("📋 Training Configuration:")
            key_params = ['epochs', 'batch', 'lr0', 'weight_decay', 'label_smoothing', 'cls', 'box']
            for key in key_params:
                if key in final_config:
                    print(f"   {key}: {final_config[key]}")
            print()
            
            print(f"⏰ Training started at: {datetime.now()}")
            
            # Try different model initialization approaches
            model = None
            model_init_strategies = [
                ('pretrained_yolov10n', lambda: YOLO('yolov10n.pt')),
                ('pretrained_yolov8n', lambda: YOLO('yolov8n.pt')),  # Fallback
                ('pretrained_yolov5n', lambda: YOLO('yolov5nu.pt')),  # Another fallback
                ('scratch_yolov8', lambda: YOLO('yolov8n.yaml'))  # From scratch
            ]
            
            for strategy_name, init_func in model_init_strategies:
                try:
                    print(f"🔄 Attempting model initialization: {strategy_name}")
                    model = init_func()
                    print(f"✅ Model initialized successfully with: {strategy_name}")
                    break
                except Exception as e:
                    print(f"❌ Failed {strategy_name}: {str(e)[:100]}...")
                    continue
            
            if model is None:
                raise RuntimeError("All model initialization strategies failed")
            
            # Create holdout test set before training (with error handling)
            print("🔍 Creating holdout test set for overfitting detection...")
            holdout_dir = None
            try:
                if self.overfitting_validator:
                    holdout_dir = self.overfitting_validator.create_holdout_test_set(test_ratio=0.15)
                else:
                    print("   ⚠️ No overfitting validator available")
            except Exception as e:
                print(f"   ⚠️ Holdout set creation failed: {e}")
            
            # Start training with error handling
            print("🏋️ Starting training...")
            start_time = datetime.now()
            
            try:
                results = model.train(**final_config)
                training_success = True
            except Exception as e:
                print(f"❌ Primary training failed: {e}")
                print("🔄 Trying simplified configuration...")
                
                # Fallback with simpler config
                simple_config = {
                    'data': str(PATHS['data_yaml']),
                    'epochs': min(50, strategy_config['epochs']),
                    'batch': max(4, strategy_config['batch'] // 2),
                    'imgsz': 640,
                    'device': str(device).replace('cuda:', ''),
                    'project': str(experiment_dir.parent),
                    'name': f"{experiment_name}_fallback",
                    'exist_ok': True,
                    'cache': False,
                    'verbose': True,
                    'amp': False,  # Disable AMP for compatibility
                    'workers': 2
                }
                
                try:
                    results = model.train(**simple_config)
                    training_success = True
                    print("✅ Fallback training succeeded!")
                except Exception as e2:
                    print(f"❌ Fallback training also failed: {e2}")
                    training_success = False
                    results = None
            
            end_time = datetime.now()
            training_duration = end_time - start_time
            
            if training_success and results is not None:
                print(f"\n✅ Training completed!")
                print(f"⏰ Duration: {training_duration}")
                
                # Evaluate model on validation set
                print("\n📊 Evaluating trained model...")
                try:
                    val_results = model.val()
                    print("✅ Validation completed!")
                except Exception as e:
                    print(f"⚠️ Validation failed: {e}")
                    val_results = None
                
                # Evaluate on holdout test set
                holdout_metrics = None
                if holdout_dir and self.overfitting_validator:
                    try:
                        print("\n🔍 Evaluating on holdout test set...")
                        holdout_metrics = self.overfitting_validator.validate_on_holdout(model, holdout_dir)
                        print("✅ Holdout evaluation completed!")
                    except Exception as e:
                        print(f"⚠️ Holdout evaluation failed: {e}")
                
                # Extract and compile metrics
                try:
                    metrics = self._extract_metrics(val_results, results)
                    metrics['training_duration'] = str(training_duration)
                    metrics['experiment_name'] = experiment_name
                    metrics['strategy'] = strategy
                    metrics['holdout_metrics'] = holdout_metrics
                    
                    # Perform overfitting analysis
                    if self.overfitting_validator:
                        try:
                            if hasattr(results, 'results_dict'):
                                train_metrics = results.results_dict
                            else:
                                train_metrics = None
                            
                            overfitting_analysis = self.overfitting_validator.detect_overfitting(
                                train_metrics=train_metrics,
                                val_metrics=metrics.get('validation', {}),
                                holdout_metrics=holdout_metrics
                            )
                            metrics['overfitting_analysis'] = overfitting_analysis
                        except Exception as e:
                            print(f"⚠️ Overfitting analysis failed: {e}")
                            metrics['overfitting_analysis'] = None
                    
                    print("✅ Metrics extraction completed!")
                    
                except Exception as e:
                    print(f"⚠️ Metrics extraction failed: {e}")
                    metrics = {
                        'training_duration': str(training_duration),
                        'experiment_name': experiment_name,
                        'strategy': strategy,
                        'status': 'completed_with_errors'
                    }
                
                # Save model and results
                try:
                    self._save_training_results(experiment_name, model, metrics, results)
                    print("✅ Results saved successfully!")
                except Exception as e:
                    print(f"⚠️ Failed to save results: {e}")
                
                return model, metrics, results
            
            else:
                print("❌ Training failed - check configuration and data paths")
                return None, {'status': 'failed', 'experiment_name': experiment_name}, None
        
        except Exception as e:
            print(f"❌ Training failed with error: {e}")
            return None, {'status': 'failed', 'error': str(e), 'experiment_name': experiment_name}, None
    
    def _extract_metrics(self, val_results, train_results):
        """Extract metrics with error handling"""
        metrics = {
            'validation': {},
            'training': {},
            'status': 'completed'
        }
        
        try:
            if val_results is not None:
                # Extract validation metrics
                if hasattr(val_results, 'results_dict'):
                    val_dict = val_results.results_dict
                    metrics['validation'] = {
                        'map50': val_dict.get('metrics/mAP50(B)', 0.0),
                        'map50_95': val_dict.get('metrics/mAP50-95(B)', 0.0),
                        'precision': val_dict.get('metrics/precision(B)', 0.0),
                        'recall': val_dict.get('metrics/recall(B)', 0.0),
                    }
                    
                    # Calculate F1 score
                    p = metrics['validation']['precision']
                    r = metrics['validation']['recall']
                    metrics['validation']['f1_score'] = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
            
            # Extract training metrics if available
            if train_results is not None and hasattr(train_results, 'results_dict'):
                train_dict = train_results.results_dict
                metrics['training'] = {
                    'final_loss': train_dict.get('train/box_loss', 0.0) + train_dict.get('train/cls_loss', 0.0),
                    'box_loss': train_dict.get('train/box_loss', 0.0),
                    'cls_loss': train_dict.get('train/cls_loss', 0.0),
                }
        
        except Exception as e:
            print(f"⚠️ Error extracting metrics: {e}")
            metrics['status'] = 'metrics_extraction_failed'
        
        return metrics
    
    def _save_training_results(self, experiment_name, model, metrics, results):
        """Save training results with error handling"""
        try:
            # Save model
            best_model_path = self.models_dir / f"{experiment_name}_best.pt"
            
            # Try to find and copy the best model
            possible_best_paths = [
                self.results_dir / experiment_name / 'weights' / 'best.pt',
                self.results_dir / f"{experiment_name}_fallback" / 'weights' / 'best.pt',
                PATHS['working'] / 'runs' / 'detect' / experiment_name / 'weights' / 'best.pt'
            ]
            
            model_saved = False
            for path in possible_best_paths:
                if path.exists():
                    shutil.copy2(path, best_model_path)
                    model_saved = True
                    break
            
            if not model_saved:
                # Save current model state
                try:
                    model.save(str(best_model_path))
                    model_saved = True
                except:
                    pass
            
            # Save metrics
            metrics_path = self.results_dir / f"{experiment_name}_metrics.json"
            with open(metrics_path, 'w') as f:
                json.dump(metrics, f, indent=2, default=str)
            
            print(f"📁 Model saved: {best_model_path if model_saved else 'Failed'}")
            print(f"📁 Metrics saved: {metrics_path}")
            
        except Exception as e:
            print(f"⚠️ Error saving results: {e}")

# Initialize training manager
print("🏋️ Initializing advanced training manager...")

# Check if required objects exist, create simple ones if not
if 'overfitting_validator' not in locals():
    print("⚠️ Creating simple overfitting validator...")
    class SimpleOverfittingValidator:
        def get_validation_config(self):
            return {}
        def create_holdout_test_set(self, test_ratio=0.15):
            return None
        def detect_overfitting(self, train_metrics, val_metrics, holdout_metrics):
            return {'status': 'not_available'}
        def validate_on_holdout(self, model, holdout_dir):
            return None
    overfitting_validator = SimpleOverfittingValidator()

training_manager = AdvancedTrainingManager(
    data_config=DATA_CONFIG,
    model_builder=model_builder,
    checkpoint_manager=None,  # Will use SimpleCheckpointManager
    overfitting_validator=overfitting_validator
)
print("✅ Training manager ready!")

print("\n🎯 Training Strategy Options:")
for strategy, config in training_manager.strategies.items():
    print(f"   {strategy.split('_')[0].title()}-optimized: {config['description']}")

# Run training with error handling
print("\n🚀 Starting precision-optimized training...")
print("⚠️  Note: Actual training will occur when this cell is executed")
print("📊 Previous stored outputs may show fake/cached results")

model, training_metrics, results = training_manager.train_model(
    strategy='precision_optimized_v1',
    experiment_name='precision_optimized_v1'
)

if model is not None:
    print(f"\n🎉 Training completed successfully!")
    print(f"📊 Validation mAP@0.5: {training_metrics.get('validation', {}).get('map50', 'N/A'):.3f}")
    print(f"🎯 F1 Score: {training_metrics.get('validation', {}).get('f1_score', 'N/A'):.3f}")
else:
    print(f"\n❌ Training failed - check logs above for details")
    print("💡 Common solutions:")
    print("   - Ensure dataset paths are correct")
    print("   - Check YAML configuration")
    print("   - Verify sufficient memory/storage")
    print("   - Try reducing batch size")
````

````python
# =============================================================================
# ANTI-OVERFITTING VALIDATION SYSTEM
# =============================================================================

class AntiOverfittingValidator:
    """Advanced validation system to detect and prevent overfitting"""
    
    def __init__(self, data_config, working_dir):
        self.data_config = data_config
        self.working_dir = Path(working_dir)
        self.validation_history = []
        self.overfitting_warnings = []
        
    def create_holdout_test_set(self, test_ratio=0.15):
        """Create a holdout test set from training data for final validation"""
        
        print("🔍 Creating holdout test set for overfitting detection...")
        
        # Fix path construction - use PATHS instead of malformed paths
        train_images_dir = PATHS['train_images']
        train_labels_dir = PATHS['train_labels']
        
        if not train_images_dir.exists():
            print(f"❌ Training images directory not found: {train_images_dir}")
            return False
            
        # Get all training images
        train_images = list(train_images_dir.glob('*.jpg')) + list(train_images_dir.glob('*.png'))
        
        if len(train_images) == 0:
            print("❌ No training images found!")
            return False
            
        # Create holdout directories
        holdout_dir = self.working_dir / 'holdout_test'
        holdout_images_dir = holdout_dir / 'images'
        holdout_labels_dir = holdout_dir / 'labels'
        
        holdout_images_dir.mkdir(parents=True, exist_ok=True)
        holdout_labels_dir.mkdir(parents=True, exist_ok=True)
        
        # Randomly select test images (stratified by class if possible)
        np.random.seed(42)
        num_holdout = int(len(train_images) * test_ratio)
        holdout_images = np.random.choice(train_images, num_holdout, replace=False)
        
        # Move selected images and labels to holdout
        moved_count = 0
        for img_path in holdout_images:
            try:
                # Move image
                shutil.move(str(img_path), str(holdout_images_dir / img_path.name))
                
                # Move corresponding label if exists
                label_path = train_labels_dir / (img_path.stem + '.txt')
                if label_path.exists():
                    shutil.move(str(label_path), str(holdout_labels_dir / label_path.name))
                
                moved_count += 1
            except Exception as e:
                print(f"⚠️ Failed to move {img_path.name}: {e}")
        
        print(f"✅ Created holdout test set: {moved_count} images")
        print(f"📁 Location: {holdout_dir}")
        
        return holdout_dir
    
    def validate_on_holdout(self, model, holdout_dir):
        """Validate trained model on holdout test set"""
        
        print("🔍 Validating model on holdout test set...")
        
        try:
            # Create temporary data.yaml for holdout validation
            holdout_yaml = holdout_dir / 'holdout_data.yaml'
            holdout_config = {
                'path': str(holdout_dir),
                'train': str(holdout_dir / 'images'),
                'val': str(holdout_dir / 'images'),  # Use same for validation
                'test': str(holdout_dir / 'images'),
                'nc': self.data_config['nc'],
                'names': self.data_config['names']
            }
            
            with open(holdout_yaml, 'w') as f:
                yaml.dump(holdout_config, f)
            
            # Run validation on holdout set
            results = model.val(data=str(holdout_yaml), split='test')
            
            # Extract holdout metrics
            holdout_metrics = {
                'precision': float(results.box.mp) if hasattr(results, 'box') else 0.0,
                'recall': float(results.box.mr) if hasattr(results, 'box') else 0.0,
                'mAP50': float(results.box.map50) if hasattr(results, 'box') else 0.0,
                'mAP50_95': float(results.box.map) if hasattr(results, 'box') else 0.0,
                'timestamp': datetime.now().isoformat()
            }
            
            # Calculate F1 score
            p, r = holdout_metrics['precision'], holdout_metrics['recall']
            holdout_metrics['f1_score'] = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
            
            print(f"📊 Holdout Test Results:")
            print(f"   Precision: {holdout_metrics['precision']:.4f}")
            print(f"   Recall: {holdout_metrics['recall']:.4f}")
            print(f"   F1-Score: {holdout_metrics['f1_score']:.4f}")
            print(f"   mAP@0.5: {holdout_metrics['mAP50']:.4f}")
            
            return holdout_metrics
            
        except Exception as e:
            print(f"❌ Holdout validation failed: {e}")
            return None
    
    def detect_overfitting(self, train_metrics, val_metrics, holdout_metrics=None):
        """Detect overfitting by comparing training, validation, and holdout metrics"""
        
        print("\n🔍 OVERFITTING ANALYSIS")
        print("=" * 50)
        
        overfitting_detected = False
        warnings = []
        
        # Compare training vs validation performance
        if train_metrics and val_metrics:
            train_acc = train_metrics.get('mAP50', 0)
            val_acc = val_metrics.get('mAP50', 0)
            
            if train_acc > val_acc:
                gap = train_acc - val_acc
                if gap > 0.1:  # 10% gap threshold
                    overfitting_detected = True
                    warnings.append(f"Large train-val gap: {gap:.3f} (train: {train_acc:.3f}, val: {val_acc:.3f})")
        
        # Compare validation vs holdout performance
        if val_metrics and holdout_metrics:
            val_acc = val_metrics.get('mAP50', 0)
            holdout_acc = holdout_metrics.get('mAP50', 0)
            
            if val_acc > holdout_acc:
                gap = val_acc - holdout_acc
                if gap > 0.05:  # 5% gap threshold for val-holdout
                    overfitting_detected = True
                    warnings.append(f"Validation-holdout gap: {gap:.3f} (val: {val_acc:.3f}, holdout: {holdout_acc:.3f})")
        
        # Check for unrealistic high performance (potential overfitting indicator)
        if val_metrics:
            val_precision = val_metrics.get('precision', 0)
            val_recall = val_metrics.get('recall', 0)
            
            if val_precision > 0.98 or val_recall > 0.98:
                warnings.append(f"Suspiciously high performance: P={val_precision:.3f}, R={val_recall:.3f}")
                overfitting_detected = True
        
        # Print analysis results
        if overfitting_detected:
            print("🚨 OVERFITTING DETECTED!")
            for warning in warnings:
                print(f"   ⚠️ {warning}")
            
            print("\n💡 Recommendations:")
            print("   🔧 Increase regularization (weight_decay, label_smoothing)")
            print("   📉 Reduce model complexity or add dropout")
            print("   🔄 Use early stopping with lower patience")
            print("   📊 Increase dataset size or improve augmentation")
            print("   🎯 Lower confidence threshold during inference")
        else:
            print("✅ No significant overfitting detected")
            print("   Model appears to generalize well")
        
        # Store analysis results
        analysis_result = {
            'overfitting_detected': overfitting_detected,
            'warnings': warnings,
            'train_metrics': train_metrics,
            'val_metrics': val_metrics,
            'holdout_metrics': holdout_metrics,
            'timestamp': datetime.now().isoformat()
        }
        
        self.validation_history.append(analysis_result)
        
        return analysis_result

# Initialize anti-overfitting validator
print("🛡️ Initializing anti-overfitting validation system...")

overfitting_validator = AntiOverfittingValidator(
    data_config=DATA_CONFIG,
    working_dir=PATHS['working']
)

print("✅ Anti-overfitting validator ready!")
print("   🔍 Holdout test set creation")
print("   📊 Train-val-holdout comparison")  
print("   🚨 Overfitting detection alerts")
print("   💡 Automatic recommendations")
````

````python
# =============================================================================
# MODEL INTERPRETABILITY AND DEBUGGING TOOLS
# =============================================================================

class ModelDebuggingTools:
    """Tools for debugging model performance and understanding failures"""
    
    def __init__(self, model, class_names):
        self.model = model
        self.class_names = class_names
        
    def analyze_prediction_confidence(self, test_images_dir, conf_threshold=0.25):
        """Analyze prediction confidence distribution"""
        
        print("🔍 Analyzing prediction confidence patterns...")
        
        test_images = list(Path(test_images_dir).glob('*.jpg')) + list(Path(test_images_dir).glob('*.png'))
        
        if len(test_images) == 0:
            print("❌ No test images found!")
            return None
            
        confidence_scores = []
        predictions_count = 0
        
        # Sample random images for analysis
        sample_images = np.random.choice(test_images, min(50, len(test_images)), replace=False)
        
        for img_path in tqdm(sample_images, desc="Analyzing predictions"):
            try:
                results = self.model(str(img_path), conf=conf_threshold, verbose=False)
                
                if results and len(results) > 0:
                    result = results[0]
                    if hasattr(result, 'boxes') and result.boxes is not None:
                        confidences = result.boxes.conf.cpu().numpy()
                        confidence_scores.extend(confidences)
                        predictions_count += len(confidences)
                        
            except Exception as e:
                print(f"⚠️ Error processing {img_path.name}: {e}")
                continue
        
        if len(confidence_scores) == 0:
            print("❌ No predictions found! Model may be overfitted or confidence threshold too high.")
            return {
                'mean_confidence': 0.0,
                'predictions_count': 0,
                'recommendation': 'Lower confidence threshold or retrain model'
            }
        
        # Analyze confidence distribution
        confidence_stats = {
            'mean_confidence': np.mean(confidence_scores),
            'median_confidence': np.median(confidence_scores),
            'std_confidence': np.std(confidence_scores),
            'min_confidence': np.min(confidence_scores),
            'max_confidence': np.max(confidence_scores),
            'predictions_count': predictions_count,
            'images_analyzed': len(sample_images)
        }
        
        print(f"📊 Confidence Analysis Results:")
        print(f"   Mean Confidence: {confidence_stats['mean_confidence']:.3f}")
        print(f"   Median Confidence: {confidence_stats['median_confidence']:.3f}")
        print(f"   Std Deviation: {confidence_stats['std_confidence']:.3f}")
        print(f"   Range: {confidence_stats['min_confidence']:.3f} - {confidence_stats['max_confidence']:.3f}")
        print(f"   Total Predictions: {predictions_count}")
        
        # Provide recommendations based on confidence patterns
        recommendations = []
        
        if confidence_stats['mean_confidence'] > 0.9:
            recommendations.append("⚠️ Very high confidence - possible overfitting")
            recommendations.append("🔧 Consider increasing regularization")
            
        elif confidence_stats['mean_confidence'] < 0.4:
            recommendations.append("⚠️ Low confidence - model uncertainty")
            recommendations.append("🔧 Consider more training or data augmentation")
            
        if confidence_stats['std_confidence'] < 0.1:
            recommendations.append("⚠️ Low confidence variance - possible overfitting")
            
        if predictions_count == 0:
            recommendations.append("❌ No predictions - model not detecting objects")
            recommendations.append("🔧 Lower confidence threshold or retrain")
        
        confidence_stats['recommendations'] = recommendations
        
        if recommendations:
            print(f"\n💡 Recommendations:")
            for rec in recommendations:
                print(f"   {rec}")
        
        return confidence_stats
    
    def test_on_external_images(self, external_images_dir, conf_threshold=0.25):
        """Test model on completely external images (out of distribution)"""
        
        print("🌐 Testing model on external/out-of-distribution images...")
        
        external_images = list(Path(external_images_dir).glob('*.jpg')) + list(Path(external_images_dir).glob('*.png'))
        
        if len(external_images) == 0:
            print("❌ No external images found!")
            return None
            
        results_summary = {
            'total_images': len(external_images),
            'images_with_predictions': 0,
            'total_predictions': 0,
            'avg_confidence': 0.0,
            'class_distribution': {name: 0 for name in self.class_names}
        }
        
        all_confidences = []
        
        sample_images = external_images[:20]  # Test on first 20 images
        
        for img_path in tqdm(sample_images, desc="Testing external images"):
            try:
                results = self.model(str(img_path), conf=conf_threshold, verbose=False)
                
                if results and len(results) > 0:
                    result = results[0]
                    if hasattr(result, 'boxes') and result.boxes is not None and len(result.boxes) > 0:
                        results_summary['images_with_predictions'] += 1
                        
                        confidences = result.boxes.conf.cpu().numpy()
                        class_ids = result.boxes.cls.cpu().numpy().astype(int)
                        
                        results_summary['total_predictions'] += len(confidences)
                        all_confidences.extend(confidences)
                        
                        # Count class predictions
                        for cls_id in class_ids:
                            if cls_id < len(self.class_names):
                                class_name = self.class_names[cls_id]
                                results_summary['class_distribution'][class_name] += 1
                                
            except Exception as e:
                print(f"⚠️ Error processing {img_path.name}: {e}")
                continue
        
        # Calculate statistics
        if all_confidences:
            results_summary['avg_confidence'] = np.mean(all_confidences)
        
        # Print results
        print(f"\n📊 External Image Testing Results:")
        print(f"   Images Tested: {len(sample_images)}")
        print(f"   Images with Predictions: {results_summary['images_with_predictions']}")
        print(f"   Total Predictions: {results_summary['total_predictions']}")
        print(f"   Average Confidence: {results_summary['avg_confidence']:.3f}")
        
        if results_summary['total_predictions'] > 0:
            print(f"   Class Distribution:")
            for class_name, count in results_summary['class_distribution'].items():
                if count > 0:
                    print(f"     {class_name}: {count}")
        
        # Overfitting analysis based on external testing
        detection_rate = results_summary['images_with_predictions'] / len(sample_images)
        
        print(f"\n🔍 Overfitting Analysis:")
        if detection_rate > 0.8:
            print("   🚨 HIGH detection rate on external images - possible false positives")
            print("   💡 Model may be overfitted - increase confidence threshold")
        elif detection_rate < 0.1:
            print("   ⚠️ VERY LOW detection rate - model may be too conservative")
            print("   💡 Consider lowering confidence threshold or retraining")
        else:
            print("   ✅ Reasonable detection rate on external images")
        
        return results_summary

# Initialize debugging tools
print("🔧 Initializing model debugging and interpretability tools...")

# Note: debugging_tools will be initialized after model training
print("✅ Debugging tools ready for post-training analysis!")
print("   📊 Confidence distribution analysis")
print("   🌐 External image testing")
print("   🔍 Overfitting detection via out-of-distribution testing")
````

````python
# =============================================================================
# ENSEMBLE TRAINING FOR >95% PERFORMANCE TARGET
# =============================================================================

class EnsembleTrainingManager:
    """Train multiple models with different strategies for ensemble"""
    
    def __init__(self, training_manager, checkpoint_manager):
        self.training_manager = training_manager
        self.checkpoint_manager = checkpoint_manager
        self.ensemble_models = {}
        self.ensemble_metrics = {}
    
    def train_ensemble_models(self):
        """Train multiple models with different configurations"""
        
        print("🎭 Starting ensemble training for >95% performance...")
        print("Training multiple models with different strategies...")
        print("=" * 60)
        
        # Define ensemble configurations
        ensemble_configs = {
            'precision_focused': {
                'strategy': 'precision_optimized_v1',
                'experiment_name': 'ensemble_precision_v1',
                'description': 'Optimized for maximum precision'
            },
            'recall_focused': {
                'strategy': 'recall_optimized_v1', 
                'experiment_name': 'ensemble_recall_v1',
                'description': 'Optimized for maximum recall'
            },
            'balanced_optimized': {
                'strategy': 'balanced_v1',
                'experiment_name': 'ensemble_balanced_v1', 
                'description': 'Optimized for F1-score balance'
            }
        }
        
        # Train each model in ensemble
        for model_type, config in ensemble_configs.items():
            print(f"\n🔄 Training {model_type} model...")
            print(f"📋 {config['description']}")
            
            try:
                model, metrics, results = self.training_manager.train_model(
                    experiment_name=config['experiment_name'],
                    strategy=config['strategy']
                )
                
                if model and metrics:
                    self.ensemble_models[model_type] = {
                        'model': model,
                        'metrics': metrics,
                        'results': results,
                        'config': config
                    }
                    print(f"✅ {model_type} training completed successfully!")
                    self._print_model_metrics(model_type, metrics)
                else:
                    print(f"❌ {model_type} training failed!")
                    
            except Exception as e:
                print(f"❌ Error training {model_type}: {e}")
                continue
        
        # Analyze ensemble performance
        self._analyze_ensemble_performance()
        
        # Create ensemble backup
        ensemble_data = {
            'models': {k: v['metrics'] for k, v in self.ensemble_models.items()},
            'ensemble_analysis': self.ensemble_metrics,
            'timestamp': datetime.now().isoformat()
        }
        
        self.checkpoint_manager.save_checkpoint(ensemble_data, "ensemble_training")
        backup_path = self.checkpoint_manager.create_backup("05_ensemble_training")
        
        print(f"\n🔐 Ensemble training backed up: {backup_path.name}")
        
        return self.ensemble_models
    
    def _print_model_metrics(self, model_type, metrics):
        """Print individual model metrics"""
        
        val_metrics = metrics.get('validation', {})
        print(f"   📊 {model_type} Performance:")
        print(f"      Precision: {val_metrics.get('precision', 0):.4f}")
        print(f"      Recall: {val_metrics.get('recall', 0):.4f}")
        print(f"      F1-Score: {val_metrics.get('f1_score', 0):.4f}")
        
        # Safe access to mAP50 with multiple possible key names
        map50_value = 0
        for key in ['mAP50', 'map50', 'ap50', 'mAP_50', 'AP50']:
            if key in val_metrics:
                map50_value = val_metrics[key]
                break
        
        print(f"      mAP@0.5: {map50_value:.4f}")
    
    def _analyze_ensemble_performance(self):
        """Analyze overall ensemble performance"""
        
        if not self.ensemble_models:
            print("⚠️ No ensemble models available for analysis")
            return
        
        print("\n📊 ENSEMBLE PERFORMANCE ANALYSIS")
        print("=" * 50)
        
        # Collect all metrics
        all_metrics = []
        for model_type, model_data in self.ensemble_models.items():
            metrics = model_data['metrics'].get('validation', {})
            metrics['model_type'] = model_type
            all_metrics.append(metrics)
        
        # Create DataFrame for analysis
        metrics_df = pd.DataFrame(all_metrics)
        
        if not metrics_df.empty:
            # Helper function to safely get metric values
            def safe_metric_stats(df, column_name, possible_names):
                """Get statistics for a metric with multiple possible column names"""
                actual_column = None
                for name in possible_names:
                    if name in df.columns:
                        actual_column = name
                        break
                
                if actual_column is not None and df[actual_column].notna().any():
                    return {
                        'mean': float(df[actual_column].mean()),
                        'max': float(df[actual_column].max()),
                        'min': float(df[actual_column].min()),
                        'std': float(df[actual_column].std())
                    }
                else:
                    return {
                        'mean': 0.0, 'max': 0.0, 'min': 0.0, 'std': 0.0
                    }
            
            # Calculate ensemble statistics with safe access
            ensemble_stats = {
                'precision': safe_metric_stats(metrics_df, 'precision', ['precision', 'p', 'prec']),
                'recall': safe_metric_stats(metrics_df, 'recall', ['recall', 'r', 'rec']),
                'f1_score': safe_metric_stats(metrics_df, 'f1_score', ['f1_score', 'f1', 'F1']),
                'mAP50': safe_metric_stats(metrics_df, 'mAP50', ['mAP50', 'map50', 'ap50', 'mAP_50', 'AP50'])
            }
            
            # Print ensemble statistics
            print("🎯 Ensemble Model Statistics:")
            for metric, stats in ensemble_stats.items():
                print(f"   {metric}:")
                print(f"      Mean: {stats['mean']:.4f} ± {stats['std']:.4f}")
                print(f"      Range: {stats['min']:.4f} - {stats['max']:.4f}")
            
            # Find best performing models with safe column access
            def safe_find_best_model(df, metric_names, metric_display_name):
                """Safely find the best performing model for a metric"""
                actual_column = None
                for name in metric_names:
                    if name in df.columns and df[name].notna().any():
                        actual_column = name
                        break
                
                if actual_column is not None:
                    best_idx = df[actual_column].idxmax()
                    best_model = df.loc[best_idx, 'model_type']
                    best_value = df.loc[best_idx, actual_column]
                    return best_model, best_value
                else:
                    return 'N/A', 0.0
            
            best_precision_model, best_precision_value = safe_find_best_model(
                metrics_df, ['precision', 'p', 'prec'], 'Precision'
            )
            best_recall_model, best_recall_value = safe_find_best_model(
                metrics_df, ['recall', 'r', 'rec'], 'Recall'
            )
            best_f1_model, best_f1_value = safe_find_best_model(
                metrics_df, ['f1_score', 'f1', 'F1'], 'F1-Score'
            )
            
            print(f"\n🏆 Best Individual Models:")
            print(f"   Precision: {best_precision_model} ({best_precision_value:.4f})")
            print(f"   Recall: {best_recall_model} ({best_recall_value:.4f})")
            print(f"   F1-Score: {best_f1_model} ({best_f1_value:.4f})")
            
            # Performance target analysis
            target_precision = 0.95
            target_recall = 0.90
            target_f1 = 0.92
            
            precision_achieved = best_precision_value >= target_precision
            recall_achieved = best_recall_value >= target_recall
            f1_achieved = best_f1_value >= target_f1
            
            print(f"\n🎯 Performance Target Achievement:")
            print(f"   >95% Precision: {'✅' if precision_achieved else '❌'} (Best: {best_precision_value:.1%})")
            print(f"   >90% Recall: {'✅' if recall_achieved else '❌'} (Best: {best_recall_value:.1%})")
            print(f"   >92% F1-Score: {'✅' if f1_achieved else '❌'} (Best: {best_f1_value:.1%})")
            
            # Store analysis results
            self.ensemble_metrics = {
                'statistics': ensemble_stats,
                'best_models': {
                    'precision': best_precision_model,
                    'recall': best_recall_model,
                    'f1_score': best_f1_model
                },
                'targets_achieved': {
                    'precision': precision_achieved,
                    'recall': recall_achieved,
                    'f1_score': f1_achieved
                },
                'recommendations': self._generate_ensemble_recommendations(ensemble_stats)
            }
        
        print("=" * 50)
    
    def _generate_ensemble_recommendations(self, ensemble_stats):
        """Generate recommendations based on ensemble performance"""
        
        recommendations = []
        
        # Check precision performance
        if ensemble_stats['precision']['max'] < 0.95:
            recommendations.append("🎯 Consider increasing classification loss weight for better precision")
            recommendations.append("🔧 Try higher confidence thresholds during inference")
            recommendations.append("📊 Implement hard negative mining")
        
        # Check recall performance  
        if ensemble_stats['recall']['max'] < 0.90:
            recommendations.append("🔍 Consider lower confidence thresholds for better recall")
            recommendations.append("⚖️ Adjust class weights to prioritize minority classes")
            recommendations.append("🎲 Increase data augmentation for underrepresented classes")
        
        # Check consistency
        if ensemble_stats['precision']['std'] > 0.05:
            recommendations.append("🎭 High precision variance - consider ensemble voting")
            
        if ensemble_stats['recall']['std'] > 0.05:
            recommendations.append("🎭 High recall variance - consider ensemble averaging")
        
        # General recommendations
        recommendations.append("🎯 Consider test-time augmentation for ensemble inference")
        recommendations.append("📈 Implement weighted ensemble based on validation performance")
        
        return recommendations

# Initialize and run ensemble training
print("🎭 Initializing ensemble training manager...")

ensemble_manager = EnsembleTrainingManager(
    training_manager=training_manager,
    checkpoint_manager=checkpoint_manager
)

# Train ensemble models
print("\n🚀 Starting ensemble training for >95% performance target...")
ensemble_models = ensemble_manager.train_ensemble_models()

# Print final ensemble analysis
if ensemble_models:
    print("\n🎉 Ensemble training completed!")
    print(f"✅ {len(ensemble_models)} models trained successfully")
    
    # Display recommendations
    if hasattr(ensemble_manager, 'ensemble_metrics') and 'recommendations' in ensemble_manager.ensemble_metrics:
        print("\n💡 Ensemble Recommendations:")
        for rec in ensemble_manager.ensemble_metrics['recommendations']:
            print(f"   {rec}")
else:
    print("❌ Ensemble training failed - no models were trained successfully")
````

````python
# =============================================================================
# COMPREHENSIVE MEDICAL AI EVALUATION SYSTEM
# =============================================================================

class MedicalAIEvaluator:
    """Advanced evaluation system for medical AI models"""
    
    def __init__(self, ensemble_models, data_config, results_dir):
        self.ensemble_models = ensemble_models
        self.data_config = data_config
        self.results_dir = Path(results_dir)
        self.class_names = data_config['names']
        self.evaluation_results = {}
        
    def evaluate_all_models(self):
        """Evaluate all ensemble models comprehensively"""
        
        print("📊 Starting comprehensive medical AI evaluation...")
        print("=" * 60)
        
        for model_type, model_data in self.ensemble_models.items():
            print(f"\n🔍 Evaluating {model_type} model...")
            
            model = model_data['model']
            
            # Detailed evaluation
            eval_results = self._evaluate_single_model(model, model_type)
            self.evaluation_results[model_type] = eval_results
            
            # Print summary
            self._print_model_evaluation_summary(model_type, eval_results)
        
        # Generate comparative analysis
        self._generate_comparative_analysis()
        
        # Create medical visualizations
        self._create_medical_visualizations()
        
        return self.evaluation_results
    
    def _evaluate_single_model(self, model, model_type):
        """Detailed evaluation of a single model"""
        
        results = {}
        
        try:
            # Validation evaluation
            val_results = model.val(data=PATHS['data_yaml'], split='val', save_json=True)
            
            # Extract detailed metrics
            if hasattr(val_results, 'box'):
                box = val_results.box
                results['overall_metrics'] = {
                    'precision': float(box.mp),
                    'recall': float(box.mr),
                    'mAP50': float(box.map50),
                    'mAP50_95': float(box.map),
                    'fitness': float(val_results.fitness) if hasattr(val_results, 'fitness') else 0.0
                }
                
                # Per-class metrics
                if hasattr(box, 'ap_class_index') and hasattr(box, 'ap'):
                    per_class_metrics = {}
                    for i, class_idx in enumerate(box.ap_class_index):
                        if int(class_idx) < len(self.class_names):
                            class_name = self.class_names[int(class_idx)]
                            p_i = float(box.p[i]) if hasattr(box, 'p') and i < len(box.p) else 0.0
                            r_i = float(box.r[i]) if hasattr(box, 'r') and i < len(box.r) else 0.0
                            f1_i = (2 * p_i * r_i / (p_i + r_i)) if (p_i + r_i) > 0 else 0.0
                            per_class_metrics[class_name] = {
                                'precision': p_i,
                                'recall': r_i,
                                'ap50': float(box.ap50[i]) if i < len(box.ap50) else 0.0,
                                'ap': float(box.ap[i]) if i < len(box.ap) else 0.0,
                                'f1_score': f1_i
                            }
                    results['per_class_metrics'] = per_class_metrics
            
            # Test set evaluation (if available)
            try:
                test_results = model.val(data=PATHS['data_yaml'], split='test', save_json=True)
                if hasattr(test_results, 'box'):
                    test_box = test_results.box
                    results['test_metrics'] = {
                        'precision': float(test_box.mp),
                        'recall': float(test_box.mr),
                        'mAP50': float(test_box.map50),
                        'mAP50_95': float(test_box.map),
                    }
            except:
                print(f"   ⚠️ Test evaluation not available for {model_type}")
                results['test_metrics'] = None
            
            # Calculate additional medical metrics
            results['medical_metrics'] = self._calculate_medical_metrics(results)
            
        except Exception as e:
            print(f"   ❌ Error evaluating {model_type}: {e}")
            results = {'error': str(e)}
        
        return results
    
    def _calculate_medical_metrics(self, results):
        """Calculate medical-specific metrics"""
        
        medical_metrics = {}
        
        if 'overall_metrics' in results:
            precision = results['overall_metrics']['precision']
            recall = results['overall_metrics']['recall']
            
            # Sensitivity (Recall) - Critical for medical diagnosis
            medical_metrics['sensitivity'] = recall
            
            # Specificity (1 - False Positive Rate)
            # Approximated from precision and prevalence
            if precision > 0:
                # Assuming balanced dataset for approximation
                specificity = (precision * recall) / (precision + recall - precision * recall)
                medical_metrics['specificity'] = max(0, min(1, specificity))
            else:
                medical_metrics['specificity'] = 0.0
            
            # F1-Score
            if precision + recall > 0:
                medical_metrics['f1_score'] = 2 * precision * recall / (precision + recall)
            else:
                medical_metrics['f1_score'] = 0.0
            
            # Balanced Accuracy
            sensitivity = medical_metrics['sensitivity']
            specificity = medical_metrics['specificity']
            medical_metrics['balanced_accuracy'] = (sensitivity + specificity) / 2
            
            # Clinical Priority Score (weighted for medical use)
            # Higher weight on sensitivity to minimize false negatives
            medical_metrics['clinical_priority_score'] = (0.7 * sensitivity + 0.3 * specificity)
        
        return medical_metrics
    
    def _print_model_evaluation_summary(self, model_type, results):
        """Print detailed evaluation summary for a model"""
        
        print(f"   📋 {model_type.upper()} EVALUATION RESULTS:")
        
        if 'error' in results:
            print(f"      ❌ Evaluation failed: {results['error']}")
            return
        
        # Overall metrics
        if 'overall_metrics' in results:
            metrics = results['overall_metrics']
            print(f"      📊 Overall Performance:")
            print(f"         Precision: {metrics['precision']:.4f}")
            print(f"         Recall: {metrics['recall']:.4f}")
            print(f"         mAP@0.5: {metrics['mAP50']:.4f}")
            print(f"         mAP@0.5:0.95: {metrics['mAP50_95']:.4f}")
        
        # Medical metrics
        if 'medical_metrics' in results:
            med_metrics = results['medical_metrics']
            print(f"      🏥 Medical Metrics:")
            print(f"         Sensitivity: {med_metrics.get('sensitivity', 0):.4f}")
            print(f"         Specificity: {med_metrics.get('specificity', 0):.4f}")
            print(f"         F1-Score: {med_metrics.get('f1_score', 0):.4f}")
            print(f"         Clinical Priority: {med_metrics.get('clinical_priority_score', 0):.4f}")
        
        # Performance targets
        if 'overall_metrics' in results:
            precision = results['overall_metrics']['precision']
            recall = results['overall_metrics']['recall']
            f1 = results['medical_metrics'].get('f1_score', 0)
            
            print(f"      🎯 Target Achievement:")
            print(f"         >95% Precision: {'✅' if precision > 0.95 else '❌'}")
            print(f"         >90% Recall: {'✅' if recall > 0.90 else '❌'}")
            print(f"         >92% F1: {'✅' if f1 > 0.92 else '❌'}")
    
    def _generate_comparative_analysis(self):
        """Generate comparative analysis across all models"""
        
        print("\n" + "="*60)
        print("🔬 COMPARATIVE MODEL ANALYSIS")
        print("="*60)
        
        # Collect metrics for comparison
        comparison_data = []
        for model_type, results in self.evaluation_results.items():
            if 'overall_metrics' in results and 'medical_metrics' in results:
                row = {
                    'Model': model_type,
                    'Precision': results['overall_metrics']['precision'],
                    'Recall': results['overall_metrics']['recall'],
                    'mAP50': results['overall_metrics']['mAP50'],
                    'F1_Score': results['medical_metrics'].get('f1_score', 0),
                    'Clinical_Priority': results['medical_metrics'].get('clinical_priority_score', 0)
                }
                comparison_data.append(row)
        
        if comparison_data:
            comparison_df = pd.DataFrame(comparison_data)
            
            print("📊 Model Performance Comparison:")
            print(comparison_df.round(4).to_string(index=False))
            
            # Find best models
            best_precision = comparison_df.loc[comparison_df['Precision'].idxmax()]
            best_recall = comparison_df.loc[comparison_df['Recall'].idxmax()]
            best_f1 = comparison_df.loc[comparison_df['F1_Score'].idxmax()]
            best_clinical = comparison_df.loc[comparison_df['Clinical_Priority'].idxmax()]
            
            print(f"\n🏆 Best Performing Models:")
            print(f"   Precision: {best_precision['Model']} ({best_precision['Precision']:.4f})")
            print(f"   Recall: {best_recall['Model']} ({best_recall['Recall']:.4f})")
            print(f"   F1-Score: {best_f1['Model']} ({best_f1['F1_Score']:.4f})")
            print(f"   Clinical Priority: {best_clinical['Model']} ({best_clinical['Clinical_Priority']:.4f})")
            
            # Overall best model recommendation
            # Weight clinical priority highest for medical AI
            best_overall = comparison_df.loc[comparison_df['Clinical_Priority'].idxmax()]
            print(f"\n🎯 RECOMMENDED MODEL FOR CLINICAL USE:")
            print(f"   Model: {best_overall['Model']}")
            print(f"   Clinical Priority Score: {best_overall['Clinical_Priority']:.4f}")
            print(f"   Precision: {best_overall['Precision']:.4f}")
            print(f"   Recall: {best_overall['Recall']:.4f}")
        
        print("="*60)
    
    def _create_medical_visualizations(self):
        """Create comprehensive medical AI visualizations"""
        
        print("\n🎨 Creating medical AI visualizations...")
        
        # Create visualization directory
        viz_dir = self.results_dir / 'visualizations'
        viz_dir.mkdir(exist_ok=True)
        
        try:
            # 1. Performance comparison radar chart
            self._create_performance_radar_chart(viz_dir)
            
            # 2. Per-class performance heatmap
            self._create_per_class_heatmap(viz_dir)
            
            # 3. Medical metrics comparison
            self._create_medical_metrics_chart(viz_dir)
            
            # 4. ROC-like performance visualization
            self._create_performance_scatter(viz_dir)
            
            print(f"   ✅ Visualizations saved to: {viz_dir}")
            
        except Exception as e:
            print(f"   ⚠️ Error creating visualizations: {e}")
    
    def _create_performance_radar_chart(self, viz_dir):
        """Create radar chart for model performance comparison"""
        
        fig = go.Figure()
        
        metrics = ['Precision', 'Recall', 'mAP50', 'F1_Score', 'Clinical_Priority']
        
        for model_type, results in self.evaluation_results.items():
            if 'overall_metrics' in results and 'medical_metrics' in results:
                values = [
                    results['overall_metrics']['precision'],
                    results['overall_metrics']['recall'],
                    results['overall_metrics']['mAP50'],
                    results['medical_metrics'].get('f1_score', 0),
                    results['medical_metrics'].get('clinical_priority_score', 0)
                ]
                
                fig.add_trace(go.Scatterpolar(
                    r=values,
                    theta=metrics,
                    fill='toself',
                    name=model_type.replace('_', ' ').title()
                ))
        
        fig.update_layout(
            polar=dict(
                radialaxis=dict(
                    visible=True,
                    range=[0, 1]
                )),
            showlegend=True,
            title="Model Performance Comparison (Medical AI Metrics)"
        )
        
        fig.write_html(viz_dir / 'performance_radar.html')
        print("      📊 Performance radar chart created")
    
    def _create_per_class_heatmap(self, viz_dir):
        """Create heatmap of per-class performance"""
        
        # Collect per-class data
        class_data = {}
        models = []
        
        for model_type, results in self.evaluation_results.items():
            if 'per_class_metrics' in results:
                models.append(model_type)
                for class_name, metrics in results['per_class_metrics'].items():
                    if class_name not in class_data:
                        class_data[class_name] = {}
                    class_data[class_name][model_type] = metrics.get('f1_score', 0)
        
        if class_data and models:
            # Create DataFrame
            heatmap_data = []
            for class_name in self.class_names:
                if class_name in class_data:
                    row = [class_data[class_name].get(model, 0) for model in models]
                    heatmap_data.append(row)
                else:
                    heatmap_data.append([0] * len(models))
            
            # Create heatmap
            fig = go.Figure(data=go.Heatmap(
                z=heatmap_data,
                x=[model.replace('_', ' ').title() for model in models],
                y=self.class_names,
                colorscale='RdYlBu_r',
                zmin=0,
                zmax=1
            ))
            
            fig.update_layout(
                title='Per-Class F1-Score Heatmap',
                xaxis_title='Models',
                yaxis_title='Fracture Classes'
            )
            
            fig.write_html(viz_dir / 'per_class_heatmap.html')
            print("      🔥 Per-class heatmap created")
    
    def _create_medical_metrics_chart(self, viz_dir):
        """Create medical-specific metrics comparison"""
        
        models = []
        sensitivity_scores = []
        specificity_scores = []
        f1_scores = []
        
        for model_type, results in self.evaluation_results.items():
            if 'medical_metrics' in results:
                models.append(model_type.replace('_', ' ').title())
                sensitivity_scores.append(results['medical_metrics'].get('sensitivity', 0))
                specificity_scores.append(results['medical_metrics'].get('specificity', 0))
                f1_scores.append(results['medical_metrics'].get('f1_score', 0))
        
        if models:
            fig = go.Figure()
            
            fig.add_trace(go.Bar(
                name='Sensitivity (Recall)',
                x=models,
                y=sensitivity_scores,
                marker_color='lightcoral'
            ))
            
            fig.add_trace(go.Bar(
                name='Specificity',
                x=models,
                y=specificity_scores,
                marker_color='lightblue'
            ))
            
            fig.add_trace(go.Bar(
                name='F1-Score',
                x=models,
                y=f1_scores,
                marker_color='lightgreen'
            ))
            
            fig.update_layout(
                title='Medical AI Performance Metrics',
                xaxis_title='Models',
                yaxis_title='Score',
                barmode='group',
                yaxis=dict(range=[0, 1])
            )
            
            fig.write_html(viz_dir / 'medical_metrics.html')
            print("      🏥 Medical metrics chart created")
    
    def _create_performance_scatter(self, viz_dir):
        """Create precision-recall scatter plot"""
        
        fig = go.Figure()
        
        for model_type, results in self.evaluation_results.items():
            if 'overall_metrics' in results:
                precision = results['overall_metrics']['precision']
                recall = results['overall_metrics']['recall']
                
                fig.add_trace(go.Scatter(
                    x=[precision],
                    y=[recall],
                    mode='markers+text',
                    name=model_type.replace('_', ' ').title(),
                    text=[model_type.replace('_', ' ').title()],
                    textposition="top center",
                    marker=dict(size=15)
                ))
        
        # Add target lines
        fig.add_hline(y=0.9, line_dash="dash", line_color="red", 
                     annotation_text="90% Recall Target")
        fig.add_vline(x=0.95, line_dash="dash", line_color="blue", 
                     annotation_text="95% Precision Target")
        
        fig.update_layout(
            title='Precision vs Recall Performance',
            xaxis_title='Precision',
            yaxis_title='Recall',
            xaxis=dict(range=[0, 1]),
            yaxis=dict(range=[0, 1])
        )
        
        fig.write_html(viz_dir / 'precision_recall_scatter.html')
        print("      📈 Precision-recall scatter plot created")

# Initialize and run comprehensive evaluation
if 'ensemble_models' in locals() and ensemble_models:
    print("📊 Starting comprehensive medical AI evaluation...")
    
    evaluator = MedicalAIEvaluator(
        ensemble_models=ensemble_models,
        data_config=DATA_CONFIG,
        results_dir=PATHS['working'] / 'results'
    )
    
    # Run comprehensive evaluation
    evaluation_results = evaluator.evaluate_all_models()
    
    # Save evaluation results
    checkpoint_manager.save_checkpoint(evaluation_results, "comprehensive_evaluation")
    backup_path = checkpoint_manager.create_backup("06_comprehensive_evaluation")
    
    print(f"\n✅ Comprehensive evaluation completed!")
    print(f"🔐 Results backed up to: {backup_path.name}")
    
else:
    print("⚠️ No ensemble models available for evaluation")
````

# 🚀 ADVANCED POST-TRAINING OPTIMIZATIONS

This section implements cutting-edge post-training techniques to maximize model performance:

## Key Post-Training Enhancements:
- **Test-Time Augmentation (TTA)**: Multiple augmented predictions for robust inference
- **Weighted Box Fusion (WBF)**: Intelligent bounding box merging from multiple models
- **Model Fusion Strategies**: Ensemble methods for superior accuracy
- **Adaptive Confidence Thresholding**: Dynamic threshold optimization
- **Medical-Grade Quality Assurance**: Confidence-based prediction filtering

````python
# =============================================================================
# ADVANCED POST-TRAINING OPTIMIZATION PIPELINE
# =============================================================================

import ensemble_boxes
from ensemble_boxes import weighted_boxes_fusion
import albumentations as A
from albumentations.pytorch import ToTensorV2

class AdvancedPostTrainingOptimizer:
    """Advanced post-training optimization techniques for medical AI"""
    
    def __init__(self, ensemble_models, data_config, confidence_threshold=0.25):
        self.ensemble_models = ensemble_models
        self.data_config = data_config
        self.class_names = data_config['names']
        self.nc = data_config['nc']
        self.conf_threshold = confidence_threshold
        
        # Test-Time Augmentation configurations
        self.tta_transforms = self._setup_tta_transforms()
        
        # Weighted Box Fusion parameters
        self.wbf_params = {
            'iou_thr': 0.5,
            'skip_box_thr': 0.01
        }
        
    def _setup_tta_transforms(self):
        """Setup Test-Time Augmentation transforms"""
        
        print("🔄 Setting up Test-Time Augmentation (TTA) transforms...")
        
        # Medical-safe TTA transforms
        tta_transforms = [
            # Original image (no transform)
            A.Compose([A.Normalize(), ToTensorV2()], p=1.0),
            
            # Horizontal flip (anatomically valid)
            A.Compose([
                A.HorizontalFlip(p=1.0),
                A.Normalize(),
                ToTensorV2()
            ], p=1.0),
            
            # Small rotation (preserve anatomy)
            A.Compose([
                A.Rotate(limit=5, p=1.0),
                A.Normalize(),
                ToTensorV2()
            ], p=1.0),
            
            # Brightness adjustment (imaging conditions)
            A.Compose([
                A.RandomBrightnessContrast(
                    brightness_limit=0.1, 
                    contrast_limit=0.1, 
                    p=1.0
                ),
                A.Normalize(),
                ToTensorV2()
            ], p=1.0),
            
            # Scale variation (zoom levels)
            A.Compose([
                A.RandomScale(scale_limit=0.1, p=1.0),
                A.PadIfNeeded(min_height=640, min_width=640, border_mode=0),
                A.CenterCrop(height=640, width=640),
                A.Normalize(),
                ToTensorV2()
            ], p=1.0),
            
            # Multi-augmentation combination
            A.Compose([
                A.HorizontalFlip(p=0.5),
                A.Rotate(limit=3, p=0.5),
                A.RandomBrightnessContrast(
                    brightness_limit=0.05, 
                    contrast_limit=0.05, 
                    p=0.5
                ),
                A.Normalize(),
                ToTensorV2()
            ], p=1.0)
        ]
        
        print(f"   ✅ {len(tta_transforms)} TTA transforms configured")
        
        return tta_transforms
    
    def apply_test_time_augmentation(self, image_path, model):
        """Apply Test-Time Augmentation for robust predictions"""
        
        all_predictions = []
        
        try:
            # Load original image
            import cv2
            image = cv2.imread(str(image_path))
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            
            # Apply each TTA transform and get predictions
            for i, transform in enumerate(self.tta_transforms):
                try:
                    # Apply transform
                    transformed = transform(image=image)
                    transformed_image = transformed['image']
                    
                    # Run inference with YOLO model
                    results = model(image_path, conf=self.conf_threshold, save=False, verbose=False)
                    
                    if results and len(results) > 0:
                        result = results[0]
                        
                        # Extract predictions
                        if hasattr(result, 'boxes') and result.boxes is not None:
                            boxes = result.boxes
                            
                            # Convert to standard format
                            if len(boxes) > 0:
                                boxes_xyxy = boxes.xyxy.cpu().numpy()  # x1, y1, x2, y2
                                confidences = boxes.conf.cpu().numpy()
                                class_ids = boxes.cls.cpu().numpy().astype(int)
                                
                                # Apply inverse transform if needed (for flips/rotations)
                                if i == 1:  # Horizontal flip
                                    # Flip boxes back
                                    img_width = image.shape[1]
                                    boxes_xyxy[:, [0, 2]] = img_width - boxes_xyxy[:, [2, 0]]
                                
                                predictions = {
                                    'boxes': boxes_xyxy,
                                    'scores': confidences,
                                    'labels': class_ids,
                                    'transform_id': i
                                }
                                
                                all_predictions.append(predictions)
                
                except Exception as e:
                    print(f"   ⚠️ TTA transform {i} failed: {e}")
                    continue
        
        except Exception as e:
            print(f"   ❌ TTA failed for {image_path}: {e}")
            return None
        
        return all_predictions
    
    def weighted_box_fusion_predictions(self, all_predictions, image_width, image_height):
        """Apply Weighted Box Fusion to combine predictions"""
        
        if not all_predictions or len(all_predictions) == 0:
            return None
        
        try:
            # Prepare data for WBF
            boxes_list = []
            scores_list = []
            labels_list = []
            weights = []
            
            for pred in all_predictions:
                if len(pred['boxes']) > 0:
                    # Normalize coordinates for WBF (expects 0-1 range)
                    norm_boxes = pred['boxes'].copy()
                    norm_boxes[:, [0, 2]] /= image_width   # x coordinates
                    norm_boxes[:, [1, 3]] /= image_height  # y coordinates
                    
                    # Convert from xyxy to xywh format for WBF
                    wbf_boxes = []
                    for box in norm_boxes:
                        x1, y1, x2, y2 = box
                        x_center = (x1 + x2) / 2
                        y_center = (y1 + y2) / 2
                        width = x2 - x1
                        height = y2 - y1
                        wbf_boxes.append([x1, y1, x2, y2])  # Keep xyxy for ensemble_boxes
                    
                    boxes_list.append(wbf_boxes)
                    scores_list.append(pred['scores'].tolist())
                    labels_list.append(pred['labels'].tolist())
                    weights.append(1.0)  # Equal weights for now
            
            if len(boxes_list) == 0:
                return None
            
            # Apply Weighted Box Fusion
            fused_boxes, fused_scores, fused_labels = weighted_boxes_fusion(
                boxes_list,
                scores_list, 
                labels_list,
                weights=weights,
                **self.wbf_params
            )
            
            if len(fused_boxes) > 0:
                # Convert back to absolute coordinates
                fused_boxes = np.array(fused_boxes)
                fused_boxes[:, [0, 2]] *= image_width   # x coordinates
                fused_boxes[:, [1, 3]] *= image_height  # y coordinates
                
                return {
                    'boxes': fused_boxes,
                    'scores': np.array(fused_scores),
                    'labels': np.array(fused_labels, dtype=int)
                }
            
        except Exception as e:
            print(f"   ❌ WBF failed: {e}")
            return None
        
        return None
    
    def ensemble_inference_with_tta_wbf(self, image_path, save_results=True):
        """Advanced ensemble inference with TTA and WBF"""
        
        print(f"🔍 Running advanced ensemble inference on: {Path(image_path).name}")
        
        # Get image dimensions
        import cv2
        image = cv2.imread(str(image_path))
        if image is None:
            print(f"   ❌ Failed to load image: {image_path}")
            return None
        
        image_height, image_width = image.shape[:2]
        
        all_model_predictions = []
        
        # Run TTA for each ensemble model
        for model_type, model_data in self.ensemble_models.items():
            print(f"   🎭 Processing {model_type} with TTA...")
            
            model = model_data['model']
            tta_predictions = self.apply_test_time_augmentation(image_path, model)
            
            if tta_predictions and len(tta_predictions) > 0:
                # Apply WBF to TTA predictions for this model
                fused_pred = self.weighted_box_fusion_predictions(
                    tta_predictions, image_width, image_height
                )
                
                if fused_pred is not None:
                    # Weight by model performance
                    model_weight = model_data.get('metrics', {}).get('validation', {}).get('f1_score', 1.0)
                    fused_pred['model_weight'] = model_weight
                    fused_pred['model_type'] = model_type
                    all_model_predictions.append(fused_pred)
        
        # Final ensemble fusion across all models
        if len(all_model_predictions) > 0:
            final_prediction = self._ensemble_model_fusion(
                all_model_predictions, image_width, image_height
            )
            
            if save_results and final_prediction is not None:
                self._save_inference_results(image_path, final_prediction)
            
            return final_prediction
        
        return None
    
    def _ensemble_model_fusion(self, model_predictions, image_width, image_height):
        """Fuse predictions from multiple models using weighted ensemble"""
        
        if not model_predictions:
            return None
        
        try:
            # Prepare for final WBF across models
            boxes_list = []
            scores_list = []
            labels_list = []
            weights = []
            
            for pred in model_predictions:
                if len(pred['boxes']) > 0:
                    # Normalize boxes for WBF
                    norm_boxes = pred['boxes'].copy()
                    norm_boxes[:, [0, 2]] /= image_width
                    norm_boxes[:, [1, 3]] /= image_height
                    
                    boxes_list.append(norm_boxes.tolist())
                    scores_list.append(pred['scores'].tolist())
                    labels_list.append(pred['labels'].tolist())
                    weights.append(pred.get('model_weight', 1.0))
            
            if len(boxes_list) == 0:
                return None
            
            # Final ensemble WBF
            final_boxes, final_scores, final_labels = weighted_boxes_fusion(
                boxes_list,
                scores_list,
                labels_list,
                weights=weights,
                iou_thr=0.6,  # Slightly higher for final fusion
                skip_box_thr=0.02
            )
            
            if len(final_boxes) > 0:
                # Convert back to absolute coordinates
                final_boxes = np.array(final_boxes)
                final_boxes[:, [0, 2]] *= image_width
                final_boxes[:, [1, 3]] *= image_height
                
                return {
                    'boxes': final_boxes,
                    'scores': np.array(final_scores),
                    'labels': np.array(final_labels, dtype=int),
                    'ensemble_type': 'TTA_WBF_Ensemble',
                    'num_models': len(model_predictions)
                }
            
        except Exception as e:
            print(f"   ❌ Ensemble fusion failed: {e}")
            return None
        
        return None
    
    def _save_inference_results(self, image_path, prediction):
        """Save inference results with visualization"""
        
        try:
            # Create results directory
            results_dir = PATHS['working'] / 'ensemble_inference_results'
            results_dir.mkdir(exist_ok=True)
            
            # Load and visualize image
            import cv2
            import matplotlib.pyplot as plt
            
            image = cv2.imread(str(image_path))
            image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
            
            fig, ax = plt.subplots(1, 1, figsize=(12, 8))
            ax.imshow(image_rgb)
            
            # Draw predictions
            for i, (box, score, label) in enumerate(zip(
                prediction['boxes'], 
                prediction['scores'], 
                prediction['labels']
            )):
                x1, y1, x2, y2 = box
                class_name = self.class_names[int(label)]
                
                # Draw bounding box
                rect = plt.Rectangle(
                    (x1, y1), x2-x1, y2-y1,
                    fill=False, color='red', linewidth=2
                )
                ax.add_patch(rect)
                
                # Add label
                ax.text(
                    x1, y1-5, f'{class_name}: {score:.3f}',
                    color='red', fontsize=10, fontweight='bold',
                    bbox=dict(boxstyle='round,pad=0.3', facecolor='white', alpha=0.7)
                )
            
            ax.set_title(f'Ensemble Prediction: {Path(image_path).name}', fontsize=14)
            ax.axis('off')
            
            # Save visualization
            output_path = results_dir / f'ensemble_{Path(image_path).stem}.png'
            plt.savefig(output_path, dpi=300, bbox_inches='tight')
            plt.close()
            
            print(f"   💾 Results saved to: {output_path}")
            
        except Exception as e:
            print(f"   ⚠️ Failed to save results: {e}")
    
    def optimize_confidence_thresholds(self, validation_images_sample=50):
        """Optimize confidence thresholds for maximum performance"""
        
        print("🎯 Optimizing confidence thresholds...")
        
        # Get validation images
        val_images_dir = PATHS['dataset'] / 'valid' / 'images'
        val_images = list(val_images_dir.glob('*.jpg'))[:validation_images_sample]
        
        threshold_range = np.arange(0.1, 0.8, 0.05)
        best_threshold = self.conf_threshold
        best_score = 0.0
        
        for threshold in threshold_range:
            print(f"   Testing threshold: {threshold:.2f}")
            
            self.conf_threshold = threshold
            scores = []
            
            for img_path in val_images[:10]:  # Quick validation
                try:
                    prediction = self.ensemble_inference_with_tta_wbf(img_path, save_results=False)
                    if prediction is not None:
                        # Simple scoring based on number of high-confidence detections
                        high_conf_detections = np.sum(prediction['scores'] > threshold)
                        scores.append(high_conf_detections)
                except:
                    continue
            
            avg_score = np.mean(scores) if scores else 0
            
            if avg_score > best_score:
                best_score = avg_score
                best_threshold = threshold
        
        self.conf_threshold = best_threshold
        print(f"   ✅ Optimal confidence threshold: {best_threshold:.2f}")
        
        return best_threshold

# Initialize post-training optimizer
print("🚀 Initializing advanced post-training optimizer...")

post_training_optimizer = AdvancedPostTrainingOptimizer(
    ensemble_models=ensemble_manager.ensemble_models,
    data_config=DATA_CONFIG,
    confidence_threshold=0.25
)

# Optimize confidence thresholds
optimal_threshold = post_training_optimizer.optimize_confidence_thresholds()

print(f"\n✅ Post-training optimizations completed!")
print(f"🎯 Optimal confidence threshold: {optimal_threshold:.2f}")
print("🎭 Ready for advanced ensemble inference with TTA + WBF!")
````

## 🎯 12. Ensemble Inference & Test-Time Augmentation

````python
# =============================================================================
# ANTI-OVERFITTING VALIDATION SYSTEM (FIXED PATHS)
# =============================================================================

class AntiOverfittingValidator:
    """Advanced validation system to detect and prevent overfitting"""
    
    def __init__(self, data_config, working_dir):
        self.data_config = data_config
        self.working_dir = Path(working_dir)
        self.validation_history = []
        self.overfitting_warnings = []
        
    def create_holdout_test_set(self, test_ratio=0.15):
        """Create a holdout test set from training data for final validation"""
        
        print("🔍 Creating holdout test set for overfitting detection...")
        
        # Use PATHS instead of malformed path construction
        train_images_dir = PATHS['train_images']
        train_labels_dir = PATHS['train_labels']
        
        if not train_images_dir.exists():
            print(f"❌ Training images directory not found: {train_images_dir}")
            return None
            
        # Get all training images
        train_images = list(train_images_dir.glob('*.jpg')) + list(train_images_dir.glob('*.png'))
        
        if len(train_images) == 0:
            print("❌ No training images found!")
            return None
            
        # Create holdout directories
        holdout_dir = self.working_dir / 'holdout_test'
        holdout_images_dir = holdout_dir / 'images'
        holdout_labels_dir = holdout_dir / 'labels'
        
        holdout_images_dir.mkdir(parents=True, exist_ok=True)
        holdout_labels_dir.mkdir(parents=True, exist_ok=True)
        
        # Randomly select test images (stratified by class if possible)
        np.random.seed(42)
        num_holdout = int(len(train_images) * test_ratio)
        holdout_images = np.random.choice(train_images, num_holdout, replace=False)
        
        # Move selected images and labels to holdout
        moved_count = 0
        for img_path in holdout_images:
            try:
                # Copy (don't move) image to preserve original training data
                shutil.copy2(str(img_path), str(holdout_images_dir / img_path.name))
                
                # Copy corresponding label if exists
                label_path = train_labels_dir / (img_path.stem + '.txt')
                if label_path.exists():
                    shutil.copy2(str(label_path), str(holdout_labels_dir / label_path.name))
                
                moved_count += 1
            except Exception as e:
                print(f"⚠️ Failed to copy {img_path.name}: {e}")
        
        print(f"✅ Created holdout test set: {moved_count} images")
        print(f"📁 Location: {holdout_dir}")
        
        return holdout_dir
    
    def validate_on_holdout(self, model, holdout_dir):
        """Validate trained model on holdout test set"""
        
        print("🔍 Validating model on holdout test set...")
        
        try:
            # Create temporary data.yaml for holdout validation
            holdout_yaml = holdout_dir / 'holdout_data.yaml'
            holdout_config = {
                'path': str(holdout_dir),
                'train': str(holdout_dir / 'images'),
                'val': str(holdout_dir / 'images'),  # Use same for validation
                'test': str(holdout_dir / 'images'),
                'nc': self.data_config['nc'],
                'names': self.data_config['names']
            }
            
            with open(holdout_yaml, 'w') as f:
                yaml.dump(holdout_config, f)
            
            # Run validation on holdout set
            results = model.val(data=str(holdout_yaml), split='test')
            
            # Extract holdout metrics
            holdout_metrics = {
                'precision': float(results.box.mp) if hasattr(results, 'box') else 0.0,
                'recall': float(results.box.mr) if hasattr(results, 'box') else 0.0,
                'mAP50': float(results.box.map50) if hasattr(results, 'box') else 0.0,
                'mAP50_95': float(results.box.map) if hasattr(results, 'box') else 0.0,
                'timestamp': datetime.now().isoformat()
            }
            
            # Calculate F1 score
            p, r = holdout_metrics['precision'], holdout_metrics['recall']
            holdout_metrics['f1_score'] = 2 * p * r / (p + r) if (p + r) > 0 else 0.0
            
            print(f"📊 Holdout Test Results:")
            print(f"   Precision: {holdout_metrics['precision']:.4f}")
            print(f"   Recall: {holdout_metrics['recall']:.4f}")
            print(f"   F1-Score: {holdout_metrics['f1_score']:.4f}")
            print(f"   mAP@0.5: {holdout_metrics['mAP50']:.4f}")
            print(f"   mAP@0.5:0.95: {holdout_metrics['mAP50_95']:.4f}")
            
            return holdout_metrics
            
        except Exception as e:
            print(f"❌ Holdout validation failed: {e}")
            return None
    
    def detect_overfitting(self, train_metrics, val_metrics, holdout_metrics):
        """Detect overfitting patterns"""
        
        analysis = {
            'status': 'healthy',
            'warnings': [],
            'recommendations': [],
            'overfitting_score': 0.0
        }
        
        # Skip analysis if metrics are incomplete
        if not train_metrics or not val_metrics:
            analysis['status'] = 'insufficient_data'
            return analysis
        
        # Check train vs validation gap
        train_acc = train_metrics.get('accuracy', train_metrics.get('f1_score', 0))
        val_acc = val_metrics.get('accuracy', val_metrics.get('f1_score', 0))
        
        if train_acc > 0 and val_acc > 0:
            gap = train_acc - val_acc
            
            if gap > 0.1:  # 10% gap
                analysis['overfitting_score'] += 0.3
                analysis['warnings'].append(f"Large train-val gap: {gap:.3f}")
                analysis['recommendations'].append("Consider stronger regularization")
            
            if gap > 0.2:  # 20% gap
                analysis['overfitting_score'] += 0.4
                analysis['status'] = 'overfitting_detected'
                analysis['recommendations'].append("Reduce model complexity or add dropout")
        
        # Check holdout performance if available
        if holdout_metrics:
            holdout_acc = holdout_metrics.get('f1_score', 0)
            if val_acc > 0 and holdout_acc > 0:
                holdout_gap = val_acc - holdout_acc
                
                if holdout_gap > 0.05:  # 5% gap to holdout
                    analysis['overfitting_score'] += 0.3
                    analysis['warnings'].append(f"Validation-holdout gap: {holdout_gap:.3f}")
        
        # Final assessment
        if analysis['overfitting_score'] > 0.5:
            analysis['status'] = 'overfitting_detected'
        elif analysis['overfitting_score'] > 0.3:
            analysis['status'] = 'overfitting_risk'
        
        return analysis
    
    def get_validation_config(self):
        """Get validation configuration"""
        return {
            'val_split': 0.2,
            'test_split': 0.1,
            'stratify': True,
            'shuffle': True,
            'random_state': 42
        }

# Initialize the corrected anti-overfitting validator
print("🛡️ Initializing anti-overfitting validation system (FIXED)...")

overfitting_validator = AntiOverfittingValidator(
    data_config=DATA_CONFIG,
    working_dir=PATHS['working']
)

print("✅ Anti-overfitting validator ready (with corrected paths)!")
print("   🔍 Holdout test set creation (fixed path issues)")
print("   📊 Train-val-holdout comparison")  
print("   🚨 Overfitting detection alerts")
print("   💡 Automatic recommendations")
````

## 📈 13. Training Curves & Overfitting Analysis

This section loads Ultralytics training logs and images to visualize training/validation curves and assess overfitting risk. It also summarizes key signals and saves an overfitting report.

What you get:
- Loss curves (train vs val)
- Precision/Recall/mAP over epochs
- Confusion matrix and PR curves (if available)
- Automatic overfitting detection with recommendations

````python
# =============================================================================
# TRAINING CURVES, METRIC TRENDS, AND OVERFITTING ANALYSIS
# =============================================================================

import pandas as pd
import matplotlib.pyplot as plt
import json
from pathlib import Path

class TrainingCurvesAnalyzer:
    def __init__(self, results_csv_path=None):
        self.findings = {}
        
        # If direct CSV path provided, use it
        if results_csv_path:
            self.results_csv = Path(results_csv_path)
            if self.results_csv.exists():
                self.latest_run = self.results_csv.parent
                print(f" Using provided results CSV: {self.results_csv}")
            else:
                print(f" Provided CSV path not found: {results_csv_path}")
                self.latest_run = None
                self.results_csv = None
        else:
            # Auto-detect training results
            self.latest_run = self._auto_detect_training_results()
            self.results_csv = self.latest_run / 'results.csv' if self.latest_run else None
        
        if self.latest_run and self.results_csv and self.results_csv.exists():
            self.plots_dir = self.latest_run
            self.output_dir = self.latest_run / 'analysis'
            self.output_dir.mkdir(exist_ok=True, parents=True)
        else:
            print(" No valid training results found")
            self.plots_dir = None
            self.output_dir = None
    
    def _auto_detect_training_results(self):
        """Auto-detect training results from common locations"""
        search_paths = [
            Path('results.csv'),
            Path('/kaggle/working/runs/detect'),
        ]
        
        for base_path in search_paths:
            if not base_path.exists():
                continue
                
            try:
                # Look for directories that contain results.csv
                for item in base_path.iterdir():
                    if item.is_dir():
                        csv_path = item / 'results.csv'
                        if csv_path.exists():
                            print(f" Auto-detected training results: {item}")
                            return item
            except Exception as e:
                print(f" Error searching {base_path}: {e}")
                continue
        
        print(" No training results auto-detected")
        return None
    
    def _get_latest_run(self):
        if not self.runs_dir.exists():
            print(f" Runs directory not found: {self.runs_dir}")
            return None
        candidates = sorted([p for p in self.runs_dir.iterdir() if p.is_dir()], key=lambda p: p.stat().st_mtime, reverse=True)
        return candidates[0] if candidates else None
    
    def load_results(self):
        if not self.latest_run:
            print(" No training run found; skipping curve analysis.")
            return None
        if not self.results_csv or not self.results_csv.exists():
            print(f" results.csv not found at {self.results_csv}; skipping curve analysis.")
            return None
        try:
            df = pd.read_csv(self.results_csv)
            print(f" Loaded training results: {len(df)} epochs")
            # Ultralytics usually includes these columns
            expected_cols = ['epoch','train/box_loss','train/cls_loss','train/dfl_loss','val/box_loss','val/cls_loss','val/dfl_loss','metrics/precision(B)','metrics/recall(B)','metrics/mAP50(B)','metrics/mAP50-95(B)']
            missing = [c for c in expected_cols if c not in df.columns]
            if missing:
                print(f" Missing expected columns: {missing}")
                print(f"📋 Available columns: {list(df.columns)}")
            self.df = df
            return df
        except Exception as e:
            print(f" Failed to read results.csv: {e}")
            return None
    
    def plot_losses(self):
        if not hasattr(self, 'df') or not self.output_dir:
            print(" Cannot plot losses: No data or output directory")
            return
        plt.figure(figsize=(12,6))
        for col in ['train/box_loss','train/cls_loss','train/dfl_loss']:
            if col in self.df:
                plt.plot(self.df['epoch'], self.df[col], label=col)
        for col in ['val/box_loss','val/cls_loss','val/dfl_loss']:
            if col in self.df:
                plt.plot(self.df['epoch'], self.df[col], label=col, linestyle='--')
        plt.title('Training vs Validation Losses')
        plt.xlabel('Epoch')
        plt.ylabel('Loss')
        plt.legend()
        out = self.output_dir / 'loss_curves.png'
        plt.savefig(out, dpi=200, bbox_inches='tight')
        plt.close()
        print(f"💾 Saved: {out}")
    
    def plot_metrics(self):
        if not hasattr(self, 'df') or not self.output_dir:
            print(" Cannot plot metrics: No data or output directory")
            return
        metrics = {
            'metrics/precision(B)': 'Precision',
            'metrics/recall(B)': 'Recall',
            'metrics/mAP50(B)': 'mAP@50',
            'metrics/mAP50-95(B)': 'mAP@50-95'
        }
        plt.figure(figsize=(12,6))
        for k, label in metrics.items():
            if k in self.df:
                plt.plot(self.df['epoch'], self.df[k], label=label)
        plt.title('Validation Metrics over Epochs')
        plt.xlabel('Epoch')
        plt.ylabel('Score')
        plt.legend()
        out = self.output_dir / 'val_metrics.png'
        plt.savefig(out, dpi=200, bbox_inches='tight')
        plt.close()
        print(f"💾 Saved: {out}")
    
    def detect_overfitting(self):
        if not hasattr(self, 'df'):
            return
        df = self.df
        findings = {}
        # Simple heuristics
        def last_n_trend(col, n=10):
            return df[col].tail(min(n, len(df))).dropna()
        
        if 'val/box_loss' in df and 'train/box_loss' in df:
            train_last = last_n_trend('train/box_loss')
            val_last = last_n_trend('val/box_loss')
            if len(train_last) > 3 and len(val_last) > 3:
                train_down = train_last.iloc[0] - train_last.iloc[-1] > 0
                val_up = val_last.iloc[-1] - val_last.iloc[0] > 0
                findings['loss_divergence'] = bool(train_down and val_up)
        
        best_map = None
        for col in ['metrics/mAP50-95(B)','metrics/mAP50(B)']:
            if col in df:
                best_map = float(df[col].max())
                break
        findings['best_map'] = best_map
        
        # Early stopping suggestion if metrics plateau
        plateau = False
        for col in ['metrics/mAP50-95(B)','metrics/mAP50(B)']:
            if col in df and len(df) >= 15:
                last = df[col].tail(10).dropna()
                if len(last) >= 5 and (last.max() - last.min()) < 0.005:
                    plateau = True
        findings['plateau'] = plateau
        
        # Recall vs Precision balance
        if 'metrics/recall(B)' in df and 'metrics/precision(B)' in df:
            pr_gap = float(abs(df['metrics/precision(B)'].iloc[-1] - df['metrics/recall(B)'].iloc[-1]))
            findings['precision_recall_gap'] = pr_gap
        
        self.findings = findings
        if self.output_dir:
            with open(self.output_dir / 'overfitting_report.json', 'w') as f:
                json.dump(findings, f, indent=2)
            print(f"💾 Saved report: {self.output_dir / 'overfitting_report.json'}")
        else:
            print(" Cannot save overfitting report: No output directory")
        
        # Console recommendations
        print("\n🔎 Overfitting Analysis:")
        if findings.get('loss_divergence'):
            print("- Potential overfitting: training loss decreasing while validation loss increases.")
        if findings.get('plateau'):
            print("- Metrics plateau detected: consider early stopping or LR reduction.")
        gap = findings.get('precision_recall_gap')
        if gap is not None and gap > 0.1:
            print("- Large Precision-Recall gap: tune conf/iou thresholds or class weights.")
        if best_map is not None:
            print(f"- Best mAP observed: {best_map:.4f}")
    
    def copy_ultralytics_plots(self):
        if not self.plots_dir or not self.output_dir:
            print(" Cannot copy plots: No plots or output directory")
            return
        # Copy standard plots if present
        for name in ['results.png','confusion_matrix.png','PR_curve.png','F1_curve.png','P_curve.png','R_curve.png']:
            src = self.plots_dir / name
            if src.exists():
                dst = self.output_dir / name
                try:
                    import shutil
                    shutil.copy(src, dst)
                    print(f"📎 Copied: {name}")
                except Exception as e:
                    print(f" Could not copy {name}: {e}")

# Run analysis - provide direct path to your results.csv
# Update this path to point to your actual training results
results_csv_path = '/kaggle/working/training_results/ensemble_precision_v1/results.csv'

# Try the direct path first, fallback to auto-detection
try:
    curves = TrainingCurvesAnalyzer(results_csv_path)
except:
    print(" Direct path failed, trying auto-detection...")
    curves = TrainingCurvesAnalyzer()

curves.load_results()
curves.plot_losses()
curves.plot_metrics()
curves.copy_ultralytics_plots()
curves.detect_overfitting()

print(" Training curves and overfitting analysis completed.")
````
